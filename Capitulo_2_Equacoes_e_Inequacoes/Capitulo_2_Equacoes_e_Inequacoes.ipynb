## 1.1 Equações do Primeiro Grau


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Uma equação do primeiro grau em uma incógnita pode ser escrita na forma<br>
    
$$
ax+b=0,
$$
    
em que $a$ e $b$ são números reais e $a\neq0$. Resolver a equação significa determinar o valor de $x$ que torna a igualdade verdadeira.
</div>


Ao resolver uma equação, podemos realizar a mesma operação nos dois membros da igualdade. O objetivo é obter uma equação equivalente na qual a incógnita esteja isolada.


#### Exemplo 1


Resolva a equação:

$$
3x-7=11.
$$


##### Solução:


Temos que

$$
\begin{aligned}
3x-7=11
&\Rightarrow 3x=11+7\\
&\Rightarrow 3x=18\\
&\Rightarrow x=\frac{18}{3}\\
&\Rightarrow x=6.
\end{aligned}
$$

Verificando na equação original:

$$
3\cdot6-7=18-7=11.
$$

Portanto,

$$
S=\{6\}.
$$


#### Exemplo 2


Resolva as seguintes equações:

a) $2(x-3)+5=x+4$

b) $\frac{2x + 1}{3} = \frac{1}{2}$

c) $3x - \frac{5}{2} = \frac{x}{2} +3$

##### Solução do item a:

Temos que

$$
\begin{aligned}
2(x-3)+5=x+4
&\Rightarrow 2x-6+5=x+4\\
&\Rightarrow 2x-1=x+4\\
&\Rightarrow x=5.
\end{aligned}
$$

Logo,

$$
S=\{5\}.
$$


##### Solução do item b:

Temos que

$$
\begin{aligned}
\frac{2x+1}{3}=\frac12
&\Rightarrow 2(2x+1)=3\\
&\Rightarrow 4x+2=3\\
&\Rightarrow 4x=1\\
&\Rightarrow x=\frac14.
\end{aligned}
$$

Logo,

$$
S=\left\{\frac14\right\}.
$$


##### Solução do item c:

Temos que

$$
\begin{aligned}
3x-\frac52=\frac{x}{2}+3
&\Rightarrow 6x-5=x+6\\
&\Rightarrow 5x=11\\
&\Rightarrow x=\frac{11}{5}.
\end{aligned}
$$

Logo,

$$
S=\left\{\frac{11}{5}\right\}.
$$


## 1.2 Equações do Segundo Grau


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Uma equação do segundo grau pode ser escrita na forma
$$
ax^2+bx+c=0,
$$
em que $a$, $b$ e $c$ são números reais e $a\neq0$.
</div>


Quando a fatoração não é imediata, podemos utilizar a fórmula resolutiva:

$$
x=\frac{-b\pm\sqrt{\Delta}}{2a},
$$

em que

$$
\Delta=b^2-4ac.
$$

O discriminante fornece informações sobre as soluções reais:

- se $\Delta>0$, existem duas soluções reais distintas;
- se $\Delta=0$, existe uma solução real dupla;
- se $\Delta<0$, não existem soluções reais.


#### Exemplo 1


Determine as soluções reais da equação:

$$
2x^2+3x-1=0.
$$


##### Solução:


Temos

$$
a=2,\qquad b=3,\qquad c=-1.
$$

Logo,

$$
\begin{aligned}
\Delta
&=b^2-4ac\\
&=3^2-4\cdot2\cdot(-1)\\
&=17,
\end{aligned}
$$

e

$$
\begin{aligned}
x
&=\frac{-3\pm\sqrt{17}}{2\cdot2}\\
&=\frac{-3\pm\sqrt{17}}{4}.
\end{aligned}
$$

Portanto,

$$
S=
\left\{
\frac{-3-\sqrt{17}}{4},
\frac{-3+\sqrt{17}}{4}
\right\}.
$$


#### Exemplo 2


Determine as soluções reais da equação:

$$
x^2-10x+25=0.
$$

##### Solução:


Temos

$$
a=1,\qquad b=-10,\qquad c=25.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^2-4ac\\
&=(-10)^2-4\cdot1\cdot25\\
&=100-100\\
&=0.
\end{aligned}
$$

Como $\Delta=0$, a equação possui duas raízes reais iguais. Pela fórmula resolutiva,

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{10\pm\sqrt{0}}{2}\\
&=\frac{10}{2}\\
&=5.
\end{aligned}
$$

Assim,

$$
x_1=x_2=5.
$$

Portanto,

$$
S=\{5\}.
$$

Observe que, neste caso, também poderíamos resolver a equação diretamente por fatoração:

$$
x^2-10x+25=(x-5)^2,
$$

de modo que

$$
(x-5)^2=0
\Rightarrow x=5.
$$

#### Exemplo 3


Determine as soluções reais da equação:

$$
x^2+4x+5=0.
$$


##### Solução:


Temos

$$
a=1,\qquad b=4,\qquad c=5.
$$

Logo,

$$
\begin{aligned}
\Delta
&=4^2-4\cdot1\cdot5\\
&=16-20\\
&=-4.
\end{aligned}
$$

Como $\Delta<0$, não existem soluções reais. Portanto,

$$
S=\varnothing.
$$


## 1.3 Equações Fracionárias Simples

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Em uma equação fracionária, a incógnita aparece em pelo menos um denominador. Antes de iniciar os cálculos, devemos determinar os valores que anulam os denominadores. Esses valores não pertencem ao domínio da equação e não podem ser aceitos como soluções.
</div>


#### Exemplo 1


Resolva a equação:

$$
\frac{x+1}{x-2}=2.
$$


##### Solução:


A condição de existência exige que o denominador seja diferente de zero:

$$
x-2\neq0 \Rightarrow x\neq2.
$$

Resolvendo a equação,

$$
\begin{aligned}
\frac{x+1}{x-2}=2
&\Rightarrow x+1=2(x-2)\\
&\Rightarrow x+1=2x-4\\
&\Rightarrow x=5.
\end{aligned}
$$

Como $5\neq2$, o valor encontrado satisfaz a condição de existência. Logo,

$$
S=\{5\}.
$$

#### Exemplo 2


Resolva a equação:

$$
\frac{x+2}{x-1}=\frac{3}{x-1}.
$$


##### Solução:


A condição de existência exige que o denominador seja diferente de zero:

$$
x-1\neq0 \Rightarrow x\neq1.
$$

Resolvendo a equação,

$$
\begin{aligned}
\frac{x+2}{x-1}=\frac{3}{x-1}
&\Rightarrow x+2=3\\
&\Rightarrow x=1.
\end{aligned}
$$

Entretanto, o valor encontrado não satisfaz a condição de existência, pois $x=1$ anula o denominador da equação original.

Portanto,

$$
S=\varnothing.
$$

##### Observação


Uma igualdade obtida após a eliminação dos denominadores pode produzir um valor que não pertence ao domínio da equação original. Por isso, a verificação das condições de existência faz parte da resolução.

Atente-se ao fato de que buscamos sempre as soluções da equação original, e não apenas das equações intermediárias obtidas durante o processo.


## 1.4 Sistemas Lineares


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Um sistema linear de duas equações e duas incógnitas pode ser escrito na forma
$$
\begin{cases}
a_1x+b_1y=c_1,\\
a_2x+b_2y=c_2.
\end{cases}
$$
A solução é o par ordenado $(x,y)$ que satisfaz simultaneamente as duas equações.
</div>


#### Exemplo 1


Resolva o sistema pelo método da adição:

$$
\begin{cases}
x+y=5,\\
x-y=1.
\end{cases}
$$


##### Solução:


Somando as duas equações:

$$
\begin{aligned}
(x+y)+(x-y)&=5+1\\
2x&=6\\
x&=3.
\end{aligned}
$$

Substituindo em $x+y=5$:

$$
3+y=5
\Rightarrow y=2.
$$

Logo,

$$
S=\{(3,2)\}.
$$

#### Exemplo 2


Resolva o sistema pelo método da substituição:

$$
\begin{cases}
2x+y=7,\\
x+y=4.
\end{cases}
$$

##### Solução:


Da segunda equação,

$$
y=4-x.
$$

Substituindo na primeira:

$$
\begin{aligned}
2x+(4-x)=7
&\Rightarrow x+4=7\\
&\Rightarrow x=3.
\end{aligned}
$$

Assim,

$$
y=4-3=1.
$$

Portanto,

$$
S=\{(3,1)\}.
$$


## Exercícios

#### Exercício 1


Resolva as equações do primeiro grau:

a) $5(2x-1)-3=2(x+4)$

b) $\displaystyle\frac{x-2}{3}+\frac{x+1}{2}=5$


##### Solução do item a:


Temos que

$$
\begin{aligned}
5(2x-1)-3=2(x+4)
&\Rightarrow 10x-8=2x+8\\
&\Rightarrow 8x=16\\
&\Rightarrow x=2.
\end{aligned}
$$

Portanto,

$$
S=\{2\}.
$$


##### Solução do item b:


Multiplicando a equação por $6$:

$$
\begin{aligned}
\frac{x-2}{3}+\frac{x+1}{2}=5
&\Rightarrow 2(x-2)+3(x+1)=30\\
&\Rightarrow 5x-1=30\\
&\Rightarrow x=\frac{31}{5}.
\end{aligned}
$$

Logo,

$$
S=\left\{\frac{31}{5}\right\}.
$$


#### Exercício 2


Resolva as equações do segundo grau:

a) $x^2-7x+12=0$

b) $3x^2-2x-1=0$


##### Solução do item a:


Fatorando:

$$
\begin{aligned}
x^2-7x+12=0
&\Rightarrow (x-3)(x-4)=0\\
&\Rightarrow x=3\quad\text{ou}\quad x=4.
\end{aligned}
$$

Logo,

$$
S=\{3,4\}.
$$


##### Solução do item b:


Temos

$$
a=3,\qquad b=-2,\qquad c=-1.
$$

Logo,

$$
\Delta=(-2)^2-4\cdot3\cdot(-1)=16.
$$

Assim,

$$
\begin{aligned}
x
&=\frac{-(-2)\pm\sqrt{16}}{2\cdot3}\\
&=\frac{2\pm4}{6},
\end{aligned}
$$

de onde

$$
x_1=1
\qquad\text{e}\qquad
x_2=-\frac13.
$$

Portanto,

$$
S=\left\{-\frac13,1\right\}.
$$


#### Exercício 3


Resolva as equações fracionárias e indique as condições de existência:

a) $\displaystyle\frac{x+3}{x-1}=4$

b) $\displaystyle\frac{x+6}{x+2}-\frac{4}{x+2} = 0$

##### Solução do item a:


A condição de existência exige que o denominador $x-1$ seja diferente de zero:

$$
x-1\neq0 \Rightarrow x\neq1.
$$

Resolvendo a equação,

$$
\begin{aligned}
\frac{x+3}{x-1}=4
&\Rightarrow x+3=4(x-1)\\
&\Rightarrow x+3=4x-4\\
&\Rightarrow 3x=7\\
&\Rightarrow x=\frac73.
\end{aligned}
$$

Como $\frac73\neq1$, o valor encontrado satisfaz a condição de existência. Portanto,

$$
S=\left\{\frac73\right\}.
$$

##### Solução do item b:


A condição de existência exige que o denominador $x+2$ seja diferente de zero:

$$
x+2\neq0 \Rightarrow x\neq-2.
$$

Resolvendo a equação,

$$
\begin{aligned}
\frac{x+6}{x+2}-\frac{4}{x+2}=0
&\Rightarrow \frac{x+2}{x+2}=0\\
&\Rightarrow 1=0,
\qquad x\neq-2.
\end{aligned}
$$

Como a última igualdade é impossível, a equação não possui solução. Portanto,

$$
S=\varnothing.
$$

#### Exercício 4


Resolva os seguintes sistemas:

a) $
\begin{cases}
5x+y=7,\\
x-y=-1.
\end{cases}
$

b) $
\begin{cases}
3x+2y=11,\\
2x+3y=10.
\end{cases}
$


##### Solução do item a:

Somando as duas equações:

$$
\begin{aligned}
(5x+y)+(x-y)&=7+(-1)\\
6x&=6\\
x&=1.
\end{aligned}
$$

Substituindo em $5x+y=7$:

$$
5+y=7
\Rightarrow y=2.
$$

Logo,

$$
S=\{(1,2)\}.
$$


##### Solução do item b:

Multiplicando a primeira equação por $3$ e a segunda por $2$:

$$
\begin{cases}
9x+6y=33,\\
4x+6y=20.
\end{cases}
$$

Subtraindo as equações:

$$
5x=13
\Rightarrow x=\frac{13}{5}.
$$

Substituindo na primeira equação:

$$
\begin{aligned}
3\cdot\frac{13}{5}+2y=11
&\Rightarrow 2y=\frac{16}{5}\\
&\Rightarrow y=\frac85.
\end{aligned}
$$

Logo,

$$
S=\left\{\left(\frac{13}{5},\frac85\right)\right\}.
$$

## 2.1 Intervalos Reais


As soluções de inequações geralmente contêm infinitos números reais. Por isso, é comum representá-las por meio de intervalos.


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Os principais casos de representação por intervalos são apresentados a seguir.
</div>

$$
\begin{array}{c|c}
\text{Desigualdade} & \text{Intervalo} \\ \hline
x>a & (a,+\infty) \\
x\geq a & [a,+\infty) \\
x<a & (-\infty,a) \\
x\leq a & (-\infty,a] \\
a<x<b & (a,b) \\
a\leq x\leq b & [a,b]
\end{array}
$$

Os parênteses indicam que o extremo não pertence ao conjunto. Os colchetes indicam que o extremo pertence ao conjunto. Os símbolos $-\infty$ e $+\infty$ sempre são acompanhados por parênteses.


Por exemplo,

$$
x\geq-3
$$

pode ser representado por

$$
[-3,+\infty).
$$


## 2.2 Inequações do Primeiro Grau


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Uma inequação é uma sentença matemática que utiliza símbolos como<br>
    
$$
<,\qquad >,\qquad \leq,\qquad \geq.
$$
    
Resolver uma inequação significa determinar todos os valores que tornam a sentença verdadeira.
</div>


#### Exemplo 1


Resolva a inequação:

$$
3x-5>7.
$$


##### Solução:


Temos que

$$
\begin{aligned}
3x-5>7
&\Rightarrow 3x>12\\
&\Rightarrow x>4.
\end{aligned}
$$

Portanto,

$$
S=(4,+\infty).
$$


#### Exemplo 2


Resolva a inequação:

$$
-2x+3\leq9.
$$


##### Solução:


Temos que

$$
-2x+3\leq9
\Rightarrow -2x\leq6.
$$

Dividindo por $-2$, invertemos o sentido da desigualdade:

$$
x\geq-3.
$$

Logo,

$$
S=[-3,+\infty).
$$


<div style="background-color:#fff7e6; color:#333333; border-left:5px solid #f0ad4e; padding:10px; border-radius:6px;">
<strong>Cuidado:</strong> ao multiplicar ou dividir os dois membros de uma inequação por um número negativo, devemos inverter o sentido da desigualdade.
</div>


## 2.3 Inequações Produto


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Em uma inequação produto, o primeiro membro é escrito como um produto de fatores. Para resolvê-la:

1. determinamos os valores que anulam cada fator;
2. dividimos a reta real em intervalos;
3. analisamos o sinal do produto em cada intervalo;
4. selecionamos os intervalos compatíveis com a desigualdade.
</div>


#### Exemplo 1


Resolva a inequação:

$$
(x-1)(x+2)\leq0.
$$


##### Solução:


Os fatores se anulam em

$$
x=-2
\qquad\text{e}\qquad
x=1.
$$

Analisando os sinais:

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x+2 & x-1 & \text{Produto} \\ \hline
x<-2 & - & - & + \\
-2<x<1 & + & - & - \\
x>1 & + & + & +
\end{array}
$$

Como a inequação admite a igualdade, incluímos também os pontos que anulam o produto. Logo,

$$
S=[-2,1].
$$


#### Exemplo 2


Resolva a inequação:

$$
(2x-3)(x+4)>0.
$$


##### Solução:


Os fatores se anulam em

$$
x=-4
\qquad\text{e}\qquad
x=\frac32.
$$

Analisando os sinais:

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x+4 & 2x-3 & \text{Produto} \\ \hline
x<-4 & - & - & + \\
-4<x<\frac32 & + & - & - \\
x>\frac32 & + & + & +
\end{array}
$$

Como a desigualdade é estrita, os pontos $-4$ e $\frac32$ não são incluídos. Portanto,

$$
S=(-\infty,-4)\cup\left(\frac32,+\infty\right).
$$


## 2.4 Inequações Quociente


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Em uma inequação quociente, analisamos o sinal do numerador e do denominador. Os valores que anulam o numerador podem pertencer ao conjunto-solução quando a desigualdade admite a igualdade. Os valores que anulam o denominador nunca pertencem ao conjunto-solução.
</div>


#### Exemplo 1


Resolva a inequação:

$$
\frac{x-3}{x+1}>0.
$$


##### Solução:


O numerador se anula em $x=3$ e o denominador em $x=-1$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x-3 & x+1 & \text{Quociente} \\ \hline
x<-1 & - & - & + \\
-1<x<3 & - & + & - \\
x>3 & + & + & +
\end{array}
$$

Como a desigualdade é estrita, $x=3$ não é incluído; $x=-1$ é excluído por anular o denominador. Logo,

$$
S=(-\infty,-1)\cup(3,+\infty).
$$


#### Exemplo 2


Resolva a inequação:

$$
\frac{x+2}{x-4}\leq0.
$$


##### Solução:


O numerador se anula em $x=-2$ e o denominador em $x=4$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x+2 & x-4 & \text{Quociente} \\ \hline
x<-2 & - & - & + \\
-2<x<4 & + & - & - \\
x>4 & + & + & +
\end{array}
$$

Incluímos $x=-2$, pois a inequação admite a igualdade, e excluímos $x=4$, pois anula o denominador. Logo,

$$
S=[-2,4).
$$


## Exercícios


#### Exercício 1


Resolva as inequações do primeiro grau:

a) $5-2x>11$

b) $3(2x-1)\leq4x+5$


##### Solução do item a:


Temos que

$$
5-2x>11
\Rightarrow -2x>6.
$$

Dividindo por $-2$, invertemos o sentido da desigualdade:

$$
x<-3.
$$

Portanto,

$$
S=(-\infty,-3).
$$


##### Solução do item b:


Temos que

$$
\begin{aligned}
3(2x-1)\leq4x+5
&\Rightarrow 6x-3\leq4x+5\\
&\Rightarrow 2x\leq8\\
&\Rightarrow x\leq4.
\end{aligned}
$$

Portanto,

$$
S=(-\infty,4].
$$


#### Exercício 2


Resolva:

a) $(x-4)(x+1)\geq0$

b) $(x-3)(x-2)\geq2$


##### Solução do item a:


Os fatores se anulam em $x=-1$ e $x=4$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x+1 & x-4 & \text{Produto} \\ \hline
x<-1 & - & - & + \\
-1<x<4 & + & - & - \\
x>4 & + & + & +
\end{array}
$$

Como a inequação admite a igualdade, incluímos também os pontos que anulam o produto. Logo,

$$
S=(-\infty,-1]\cup[4,+\infty).
$$


##### Solução do item b:


Temos que

$$
\begin{aligned}
(x-3)(x-2)\geq2
&\Rightarrow x^2-5x+6\geq2\\
&\Rightarrow x^2-5x+4\geq0\\
&\Rightarrow (x-1)(x-4)\geq0.
\end{aligned}
$$

Os fatores se anulam em $x=1$ e $x=4$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x-1 & x-4 & \text{Produto} \\ \hline
x<1 & - & - & + \\
1<x<4 & + & - & - \\
x>4 & + & + & +
\end{array}
$$

Como a inequação admite a igualdade, incluímos também os pontos que anulam o produto. Portanto,

$$
S=(-\infty,1]\cup[4,+\infty).
$$


#### Exercício 3

Resolva:

a) $\displaystyle\frac{x+2}{x-3}\leq0$

b) $\frac{x^{2}-1}{x^{2}+2x+1} > 0$

##### Solução do item a:

O numerador se anula em $x=-2$ e o denominador em $x=3$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x+2 & x-3 & \text{Quociente} \\ \hline
x<-2 & - & - & + \\
-2<x<3 & + & - & - \\
x>3 & + & + & +
\end{array}
$$

Incluímos $x=-2$, pois a inequação admite a igualdade, e excluímos $x=3$, pois anula o denominador. Logo,

$$
S=[-2,3).
$$


##### Solução do item b:

Fatorando o numerador e o denominador:

$$
\frac{x^2-1}{x^2+2x+1}
=
\frac{(x-1)(x+1)}{(x+1)^2},
\qquad x\neq-1.
$$

Assim, para $x\neq-1$,

$$
\frac{x-1}{x+1}>0.
$$

Os pontos críticos são $x=-1$ e $x=1$.

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x-1 & x+1 & \text{Quociente} \\ \hline
x<-1 & - & - & + \\
-1<x<1 & - & + & - \\
x>1 & + & + & +
\end{array}
$$

O valor $x=-1$ é excluído por anular o denominador, e $x=1$ não é incluído porque a desigualdade é estrita. Portanto,

$$
S=(-\infty,-1)\cup(1,+\infty).
$$
