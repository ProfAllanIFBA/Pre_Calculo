# Teste de Aprendizagem da Aula 1

#### Questão 1

A solução da equação

$$
\frac{3x-2}{3}+1=\frac{x+1}{2}
$$

é:

a) $\displaystyle x=\frac13$

b) $\displaystyle x=-\frac13$

c) $\displaystyle x=\frac23$

d) $\displaystyle x=1$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Multiplicando a equação por $6$, temos

$$
\begin{aligned}
\frac{3x-2}{3}+1=\frac{x+1}{2}
&\Rightarrow 2(3x-2)+6=3(x+1)\\
&\Rightarrow 6x-4+6=3x+3\\
&\Rightarrow 6x+2=3x+3\\
&\Rightarrow 3x=1\\
&\Rightarrow x=\frac13.
\end{aligned}
$$

Portanto, a alternativa correta é a).

#### Questão 2

O conjunto-solução da equação

$$
x(2x-5)=3
$$

é:

a) $\{-3,1\}$

b) $\left\{-\frac12,3\right\}$

c) $\left\{\frac12,3\right\}$

d) $\{-1,3\}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Inicialmente,

$$
\begin{aligned}
x(2x-5)=3
&\Rightarrow 2x^2-5x=3\\
&\Rightarrow 2x^2-5x-3=0.
\end{aligned}
$$

Temos

$$
a=2,\qquad b=-5,\qquad c=-3.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^2-4ac\\
&=(-5)^2-4\cdot2\cdot(-3)\\
&=25+24\\
&=49.
\end{aligned}
$$

Pela fórmula resolutiva,

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{5\pm7}{4}.
\end{aligned}
$$

Assim,

$$
x_1=3
\qquad\text{e}\qquad
x_2=-\frac12.
$$

Portanto,

$$
S=\left\{-\frac12,3\right\}.
$$

A alternativa correta é b).

#### Questão 3

A solução do sistema

$$
\begin{cases}
x+y=7,\\
2x-6y=2
\end{cases}
$$

é o par ordenado:

a) $\displaystyle (5,2)$

b) $\displaystyle (4,3)$

c) $\displaystyle \left(\frac{3}{2},\frac{11}{2}\right)$

d) $\displaystyle \left(\frac{11}{2},\frac{3}{2}\right)$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Multiplicando a primeira equação por $-2$:

$$
\begin{cases}
-2x-2y=-14,\\
2x-6y=2.
\end{cases}
$$

Somando as equações:

$$
\begin{aligned}
-8y&=-12\\
y&=\frac32.
\end{aligned}
$$

Substituindo em $x+y=7$:

$$
\begin{aligned}
x+\frac32&=7\\
x&=\frac{11}{2}.
\end{aligned}
$$

Logo,

$$
S=\left\{\left(\frac{11}{2},\frac32\right)\right\}.
$$

A alternativa correta é d).

#### Questão 4

O conjunto-solução da equação

$$
(x-1)^2=x-1
$$

é:

a) $\{1,2\}$

b) $\{2\}$

c) $\{1\}$

d) $\{0,1\}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Temos que

$$
\begin{aligned}
(x-1)^2=x-1
&\Rightarrow (x-1)^2-(x-1)=0\\
&\Rightarrow (x-1)\big[(x-1)-1\big]=0\\
&\Rightarrow (x-1)(x-2)=0.
\end{aligned}
$$

Assim,

$$
x=1
\qquad\text{ou}\qquad
x=2.
$$

Portanto,

$$
S=\{1,2\}.
$$

A alternativa correta é a).

#### Questão 5

A solução do sistema

$$
\begin{cases}
x+\dfrac{y}{2}-1=3,\\[6pt]
\dfrac{2}{3}x-3y+4=0
\end{cases}
$$

é o par ordenado:

a) $(3,1)$

b) $(2,3)$

c) $(4,2)$

d) $(3,2)$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Organizando as equações,

$$
\begin{cases}
x+\dfrac{y}{2}-1=3,\\[6pt]
\dfrac{2}{3}x-3y+4=0
\end{cases}
\Rightarrow
\begin{cases}
2x+y-2=6,\\
2x-9y+12=0
\end{cases}
\Rightarrow
\begin{cases}
2x+y=8,\\
2x-9y=-12.
\end{cases}
$$

Subtraindo a segunda equação da primeira:

$$
\begin{aligned}
10y&=20\\
y&=2.
\end{aligned}
$$

Substituindo em $2x+y=8$:

$$
\begin{aligned}
2x+2&=8\\
2x&=6\\
x&=3.
\end{aligned}
$$

Logo,

$$
S=\{(3,2)\}.
$$

A alternativa correta é d).

# Teste de Aprendizagem da Aula 2

#### Questão 1

O conjunto-solução da inequação

$$
2x + 3 > 8 - \frac{x}{2}
$$

é:

a) $(2,+\infty)$

b) $[2,+\infty)$

c) $(-\infty,2)$

d) $(-\infty,2]$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Temos que

$$
\begin{aligned}
2x+3>8-\frac{x}{2}
&\Rightarrow 4x+6>16-x\\
&\Rightarrow 5x>10\\
&\Rightarrow x>2.
\end{aligned}
$$

Portanto,

$$
S=(2,+\infty).
$$

A alternativa correta é a).

#### Questão 2

O conjunto-solução da inequação

$$
x^{2}-2x+1<0
$$

é:

a) $(-\infty,1)\cup(1,+\infty)$

b) $\{1\}$

c) $0$

d) $\mathbb{R}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Fatorando o primeiro membro, temos

$$
x^2-2x+1=(x-1)^2.
$$

Assim, a inequação pode ser escrita como

$$
(x-1)^2<0.
$$

Como o quadrado de um número real nunca é negativo, não existe valor real de $x$ que satisfaça a inequação.

Portanto,

$$
S=\varnothing.
$$

A alternativa correta é e).

#### Questão 3

O conjunto-solução da inequação

$$
\frac{x-3}{4-2x}>0
$$

é:

a) $(2,3]$

b) $[2,3)$

c) $(-\infty,2)\cup(3,+\infty)$

d) $(2,3)$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

O numerador se anula em $x=3$ e o denominador em $x=2$.

Analisando os sinais:

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x-3 & 4-2x & \text{Quociente} \\ \hline
x<2 & - & + & - \\
2<x<3 & - & - & + \\
x>3 & + & - & -
\end{array}
$$

Como buscamos os valores para os quais o quociente é positivo, obtemos

$$
2<x<3.
$$

O valor $x=2$ é excluído por anular o denominador, e $x=3$ não é incluído porque a desigualdade é estrita. Portanto,

$$
S=(2,3).
$$

A alternativa correta é d).

#### Questão 4

O conjunto-solução da inequação

$$
\frac{(x-1)(x+2)}{x-3}\leq0
$$

é:

a) $[-2,1]$

b) $(-\infty,-2]\cup[1,3)$

c) $[-2,1]\cup(3,+\infty)$

d) $(-\infty,-2)\cup(1,3]$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Os pontos críticos são

$$
x=-2,\qquad x=1,\qquad x=3.
$$

Os valores $x=-2$ e $x=1$ anulam o numerador, enquanto $x=3$ anula o denominador.

$$
\begin{array}{c|c|c|c|c}
\text{Intervalo} & x+2 & x-1 & x-3 & \text{Quociente} \\ \hline
x<-2 & - & - & - & - \\
-2<x<1 & + & - & - & + \\
1<x<3 & + & + & - & - \\
x>3 & + & + & + & +
\end{array}
$$

Como a inequação admite a igualdade, incluímos $x=-2$ e $x=1$. O valor $x=3$ é excluído por anular o denominador. Portanto,

$$
S=(-\infty,-2]\cup[1,3).
$$

A alternativa correta é b).

#### Questão 5

O conjunto-solução da inequação

$$
(x-1)^2>3x-5
$$

é:

a) $(-\infty,2]\cup[3,+\infty)$

b) $(2,3)$

c) $(-\infty,2)\cup(3,+\infty)$

d) $[2,3]$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Inicialmente,

$$
\begin{aligned}
(x-1)^2>3x-5
&\Rightarrow x^2-2x+1>3x-5\\
&\Rightarrow x^2-5x+6>0\\
&\Rightarrow (x-2)(x-3)>0.
\end{aligned}
$$

Os fatores se anulam em

$$
x=2
\qquad\text{e}\qquad
x=3.
$$

Analisando os sinais:

$$
\begin{array}{c|c|c|c}
\text{Intervalo} & x-2 & x-3 & \text{Produto} \\ \hline
x<2 & - & - & + \\
2<x<3 & + & - & - \\
x>3 & + & + & +
\end{array}
$$

Como buscamos os valores para os quais o produto é positivo e a desigualdade é estrita, os pontos $x=2$ e $x=3$ não são incluídos. Portanto,

$$
S=(-\infty,2)\cup(3,+\infty).
$$

A alternativa correta é c).
