# Teste Diagnóstico — Semana 1

Conteúdos: propriedades aritméticas, ordem das operações, frações e manipulação algébrica básica.

#### Questão 1

Determine o valor da expressão

$$
\frac{
\left[
\frac{24-3^2}{5}+\sqrt{49}
\right]^2-2^6
}{
3^2-5
}.
$$

a) $6$

b) $8$

c) $9$

d) $12$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Resolvemos inicialmente as potências e a raiz:

$$
3^2=9,\qquad \sqrt{49}=7
\qquad\text{e}\qquad
2^6=64.
$$

Assim,

$$
\begin{aligned}
\frac{
\left[
\frac{24-3^2}{5}+\sqrt{49}
\right]^2-2^6
}{
3^2-5
}
&=
\frac{
\left[
\frac{24-9}{5}+7
\right]^2-64
}{
9-5
}\\
&=
\frac{
\left[
\frac{15}{5}+7
\right]^2-64
}{4}\\
&=
\frac{(3+7)^2-64}{4}\\
&=
\frac{10^2-64}{4}\\
&=
\frac{100-64}{4}\\
&=
\frac{36}{4}\\
&=9.
\end{aligned}
$$

Portanto, a alternativa correta é c).

#### Questão 2

Determine o valor da expressão

$$
\frac{
\displaystyle\frac56-\frac7{15}
}{
\displaystyle\frac34+\frac16
}.
$$

a) $\displaystyle\frac25$

b) $\displaystyle\frac35$

c) $\displaystyle\frac56$

d) $\displaystyle\frac52$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Calculamos separadamente o numerador e o denominador da fração principal.

No numerador,

$$
\begin{aligned}
\frac56-\frac7{15}
&=\frac{25}{30}-\frac{14}{30}\\
&=\frac{11}{30}.
\end{aligned}
$$

No denominador,

$$
\begin{aligned}
\frac34+\frac16
&=\frac{9}{12}+\frac{2}{12}\\
&=\frac{11}{12}.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
\frac{
\displaystyle\frac56-\frac7{15}
}{
\displaystyle\frac34+\frac16
}
&=
\frac{\frac{11}{30}}{\frac{11}{12}}\\
&=
\frac{11}{30}\cdot\frac{12}{11}\\
&=
\frac{12}{30}\\
&=
\frac25.
\end{aligned}
$$

Logo, a alternativa correta é a).

#### Questão 3

Determine o valor da expressão

$$
\frac{1}{
2+\displaystyle\frac{1}{
1+\frac13
}
}.
$$

a) $\displaystyle\frac{11}{4}$

b) $\displaystyle\frac8{11}$

c) $\displaystyle\frac7{11}$

d) $\displaystyle\frac{4}{11}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Em uma fração aninhada, começamos pelas operações mais internas.

Primeiramente,

$$
1+\frac13
=
\frac43.
$$

Logo,

$$
\frac{1}{1+\frac13}
=
\frac{1}{\frac43}
=
\frac34.
$$

Assim,

$$
\begin{aligned}
\frac{1}{
2+\displaystyle\frac{1}{
1+\frac13
}
}
&=
\frac{1}{2+\frac34}\\
&=
\frac{1}{\frac84+\frac34}\\
&=
\frac{1}{\frac{11}{4}}\\
&=
\frac4{11}.
\end{aligned}
$$

Portanto, a alternativa correta é d).

#### Questão 4

A forma simplificada da expressão

$$
3(2a-b)-2(a+4b)+5(b-a)
$$

é:

a) $a-6b$

b) $-a-6b$

c) $-a+6b$

d) $9a-6b$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Aplicando a propriedade distributiva, temos:

$$
\begin{aligned}
3(2a-b)-2(a+4b)+5(b-a)
&=6a-3b-2a-8b+5b-5a.
\end{aligned}
$$

Agora, reunimos os termos semelhantes:

$$
\begin{aligned}
6a-3b-2a-8b+5b-5a
&=(6a-2a-5a)+(-3b-8b+5b)\\
&=-a-6b.
\end{aligned}
$$

Portanto, a alternativa correta é b).

#### Questão 5

Considere a expressão

$$
\frac{
2ax+2ay+3bx+3by
}{
x^2+xy
}.
$$

Sua forma simplificada, acompanhada das condições necessárias para que a expressão original esteja definida, é:

a) $\displaystyle\frac{2a+3b}{x}$, com apenas $x\neq0$.

b) $\displaystyle\frac{2a+3b}{x+y}$, com $x\neq0$ e $x\neq-y$.

c) $\displaystyle\frac{2a+3b}{x}$, com $x\neq0$ e $x\neq-y$.

d) $\displaystyle\frac{2a+3b}{x+y}$, com apenas $x\neq-y$.

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Fatorando o numerador por agrupamento, temos:

$$
\begin{aligned}
2ax+2ay+3bx+3by
&=2a(x+y)+3b(x+y)\\
&=(2a+3b)(x+y).
\end{aligned}
$$

No denominador,

$$
x^2+xy=x(x+y).
$$

Portanto,

$$
\begin{aligned}
\frac{
2ax+2ay+3bx+3by
}{
x^2+xy
}
&=
\frac{(2a+3b)(x+y)}{x(x+y)}\\
&=
\frac{2a+3b}{x}.
\end{aligned}
$$

Entretanto, devemos preservar as restrições da expressão original. Como

$$
x(x+y)\neq0,
$$

devemos ter

$$
x\neq0
\qquad\text{e}\qquad
x+y\neq0.
$$

Equivalentemente,

$$
x\neq0
\qquad\text{e}\qquad
x\neq-y.
$$

Logo,

$$
\frac{
2ax+2ay+3bx+3by
}{
x^2+xy
}
=
\frac{2a+3b}{x},
$$

desde que

$$
x\neq0
\qquad\text{e}\qquad
x\neq-y.
$$

Portanto, a alternativa correta é c).
