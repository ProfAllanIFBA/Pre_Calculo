# Teste Diagnóstico — Semana 2

Conteúdos: produtos notáveis, fatoração, potenciação, radiciação e racionalização de denominadores.

#### Questão 1

Considere a expressão

$$
\frac{2x^2-7x+3}{x^2-4x+3}.
$$

Sua forma simplificada, acompanhada das condições necessárias para que a expressão original esteja definida, é:

a) $\displaystyle\frac{2x-1}{x-1}$, com apenas $x\neq1$

b) $\displaystyle\frac{2x-1}{x-3}$, com $x\neq1$ e $x\neq3$

c) $\displaystyle\frac{2x+1}{x-1}$, com $x\neq1$ e $x\neq3$

d) $\displaystyle\frac{2x-1}{x-1}$, com $x\neq1$ e $x\neq3$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Fatorando o numerador,

$$
2x^2-7x+3=(2x-1)(x-3).
$$

Fatorando o denominador,

$$
x^2-4x+3=(x-1)(x-3).
$$

Portanto,

$$
\frac{2x^2-7x+3}{x^2-4x+3}
=
\frac{(2x-1)(x-3)}{(x-1)(x-3)}
=
\frac{2x-1}{x-1}.
$$

Entretanto, devemos preservar as condições de existência da expressão original. Como

$$
(x-1)(x-3)\neq0,
$$

temos

$$
x\neq1
\qquad\text{e}\qquad
x\neq3.
$$

Logo,

$$
\frac{2x-1}{x-1},
\qquad x\neq1\text{ e }x\neq3.
$$

A alternativa correta é d).

#### Questão 2

A simplificação da expressão a seguir

$$
\frac{x^2-9}{x^2-6x+9}
$$

é dada por:

a) $\displaystyle\frac{x+3}{x-3}$, com $x\neq-3,\;3$

b) $\displaystyle\frac{x+3}{x-3}$, com apenas $x\neq3$

c) $\displaystyle\frac{x-3}{x+3}$, com $x\neq-3,\;3$

d) $\displaystyle\frac{x+3}{x+3}$, com $x\neq-3$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

O numerador é uma diferença de dois quadrados:

$$
x^2-9=(x-3)(x+3).
$$

O denominador é um trinômio quadrado perfeito:

$$
x^2-6x+9=(x-3)^2.
$$

Assim,

$$
\frac{x^2-9}{x^2-6x+9}
=
\frac{(x-3)(x+3)}{(x-3)^2}
=
\frac{x+3}{x-3}.
$$

A expressão original não está definida quando

$$
(x-3)^2=0,
$$

ou seja,

$$
x=3.
$$

Portanto,

$$
\frac{x+3}{x-3},
\qquad x\neq3.
$$

A alternativa correta é b).

#### Questão 3

Considerando $x\neq3$, a expressão a seguir

$$
\frac{x^3-27}{x-3}-(x+3)^2
$$

é equivalente a:

a) $-3x^{2}+9x$

b) $x^2-3x$

c) $-3x$

d) $3x$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Utilizando a diferença de cubos,

$$
x^3-27=x^3-3^3=(x-3)(x^2+3x+9).
$$

Como $x\neq3$,

$$
\frac{x^3-27}{x-3}
=
x^2+3x+9.
$$

Além disso,

$$
(x+3)^2=x^2+6x+9.
$$

Portanto,

$$
\begin{aligned}
\frac{x^3-27}{x-3}-(x+3)^2
&=x^2+3x+9-(x^2+6x+9)\\
&=x^2+3x+9-x^2-6x-9\\
&=-3x.
\end{aligned}
$$

Logo,

$$
-3x.
$$

A alternativa correta é c).

#### Questão 4

A expressão

$$
2\sqrt{50}-\sqrt8+3\sqrt{18}
$$

corresponde a:

a) $12\sqrt2$

b) $15\sqrt2$

c) $17\sqrt2$

d) $19\sqrt2$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Simplificando cada radical,

$$
\sqrt{50}
=
\sqrt{25\cdot2}
=
5\sqrt2,
$$

$$
\sqrt8
=
\sqrt{4\cdot2}
=
2\sqrt2,
$$

e

$$
\sqrt{18}
=
\sqrt{9\cdot2}
=
3\sqrt2.
$$

Substituindo,

$$
\begin{aligned}
2\sqrt{50}-\sqrt8+3\sqrt{18}
&=2(5\sqrt2)-2\sqrt2+3(3\sqrt2)\\
&=10\sqrt2-2\sqrt2+9\sqrt2\\
&=17\sqrt2.
\end{aligned}
$$

Logo,

$$
17\sqrt2.
$$

A alternativa correta é c).

#### Questão 5

A expressão

$$
\frac{x-1}{\sqrt{x}-1}
$$

é equivalente a:

a) $\sqrt{x}+1$, com $x\geq0$ e $x\neq1$

b) $\sqrt{x}+1$, com $x>1$

c) $\sqrt{x}-1$, com $x\geq0$ e $x\neq1$

d) $x+1$, com $x\geq0$ e $x\neq1$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Para que a expressão esteja definida, é necessário que

$$
x\geq0.
$$

Além disso, o denominador não pode ser igual a zero:

$$
\sqrt{x}-1\neq0.
$$

Logo,

$$
\sqrt{x}\neq1,
$$

e, portanto,

$$
x\neq1.
$$

Agora, multiplicamos numerador e denominador pelo conjugado do denominador:

$$
\frac{x-1}{\sqrt{x}-1}
\cdot
\frac{\sqrt{x}+1}{\sqrt{x}+1}.
$$

Assim,

$$
\frac{(x-1)(\sqrt{x}+1)}
{(\sqrt{x}-1)(\sqrt{x}+1)}.
$$

No denominador, temos uma diferença de dois quadrados:

$$
(\sqrt{x}-1)(\sqrt{x}+1)=x-1.
$$

Portanto,

$$
\frac{(x-1)(\sqrt{x}+1)}{x-1}.
$$

Como $x\neq1$, podemos simplificar o fator $x-1$:

$$
\sqrt{x}+1.
$$

Logo,

$$
\sqrt{x}+1,
\qquad x\geq0\text{ e }x\neq1.
$$

A alternativa correta é a).
