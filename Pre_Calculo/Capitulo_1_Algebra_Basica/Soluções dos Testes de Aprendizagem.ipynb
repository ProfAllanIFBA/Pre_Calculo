# Teste de Aprendizagem da Aula 1

#### Questão 1

Assinale a igualdade que não é verdadeira:

a) $6+(4+9)=(6+4)+9$

b) $3\cdot(5-2)=3\cdot5-3\cdot2$

c) $24\div(6\div2)=(24\div6)\div2$

d) $7\cdot4=4\cdot7$

e) Todas as igualdades anteriores são verdadeiras.

##### Solução:

Analisando cada igualdade, temos:

Alternativa a)

$$
\begin{aligned}
6+(4+9)
&=6+13\
&=19,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
(6+4)+9
&=10+9\
&=19.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade associativa da adição.

Alternativa b)

$$
\begin{aligned}
3\cdot(5-2)
&=3\cdot3\
&=9,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
3\cdot5-3\cdot2
&=15-6\
&=9.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade distributiva.

Alternativa c)

$$
\begin{aligned}
24\div(6\div2)
&=24\div3\
&=8,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
(24\div6)\div2
&=4\div2\
&=2.
\end{aligned}
$$

Portanto,

$$
24\div(6\div2)\neq(24\div6)\div2.
$$

Logo, a igualdade não é verdadeira, pois a divisão não é uma operação associativa.

Alternativa d)

$$
\begin{aligned}
7\cdot4&=28,\
4\cdot7&=28.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade comutativa da multiplicação.

Assim, a alternativa correta é c).

#### Questão 2

Determine o valor da expressão:

$$
\frac{\left[(18-2^3)\cdot2+\sqrt{49}\right]^2-3^4}{2^3-2}
$$

a) $96$

b) $108$

c) $176$

d) $216$

e) Nenhuma das alternativas anteriores apresenta o valor da expressão.

##### Solução:

Temos que

$$
\begin{aligned}
\frac{\left[(18-2^3)\cdot2+\sqrt{49}\right]^2-3^4}{2^3-2}
&=\frac{\left[(18-8)\cdot2+7\right]^2-81}{8-2}\\
&=\frac{\left[10\cdot2+7\right]^2-81}{6}\\
&=\frac{(20+7)^2-81}{6}\\
&=\frac{27^2-81}{6}\\
&=\frac{729-81}{6}\\
&=\frac{648}{6}\\
&=108.
\end{aligned}
$$

Assim, a alternativa correta é b).

#### Questão 3

Determine o valor da expressão:

$$
\left[\frac{\left(4\cdot(7-3)+2\right)^2}{3^2}-\sqrt{100}\right]\div2
$$

a) $10$

b) $11$

c) $12$

d) $13$

e) Nenhuma das alternativas anteriores apresenta o valor da expressão.

##### Solução:

Temos que

$$
\begin{aligned}
\left[\frac{\left(4\cdot(7-3)+2\right)^2}{3^2}-\sqrt{100}\right]\div2
&=
\left[\frac{\left(4\cdot4+2\right)^2}{9}-10\right]\div2\\
&=
\left[\frac{(16+2)^2}{9}-10\right]\div2\\
&=
\left[\frac{18^2}{9}-10\right]\div2\\
&=
\left[\frac{324}{9}-10\right]\div2\\
&=(36-10)\div2\\
&=26\div2\\
&=13.
\end{aligned}
$$

Assim, a alternativa correta é d).

#### Questão 4

Sem utilizar calculadora, determine o valor da expressão:

$$
\sqrt{
\frac{
2026\cdot(2025+2024)-2024\cdot(2026+2025)
}{2}
}
$$

a) $25$

b) $32$

c) $44$

d) $50$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Aplicando a propriedade distributiva, temos:

$$
\begin{aligned}
&2026\cdot(2025+2024)-2024\cdot(2026+2025)\\
&=2026\cdot2025+2026\cdot2024
-2024\cdot2026-2024\cdot2025.
\end{aligned}
$$

Pela propriedade comutativa da multiplicação,

$$
2026\cdot2024=2024\cdot2026.
$$

Assim, esses dois termos se anulam:

$$
\begin{aligned}
&2026\cdot2025-2024\cdot2025\\
&=2025\cdot(2026-2024)\\
&=2025\cdot2.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
\sqrt{\frac{2026\cdot(2025+2024)-2024\cdot(2026+2025)}{2}}
&=\sqrt{\frac{2025\cdot2}{2}}\\
&=\sqrt{2025}\\
&=45.
\end{aligned}
$$

Logo, a alternativa correta é e).

# Teste de Aprendizagem da Aula 2

#### Questão 1

O valor da expressão

$$
\frac{1}{2}-\frac{1}{2^{2}}+\frac{1}{2^{3}}
$$

é:

a) $1$

b) $\frac{3}{8}$

c) $\frac{3}{4}$

d) $\frac{4}{3}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Calculamos inicialmente as potências:

$$
2^2=4
\ \ \text{e} \ \
2^3=8.
$$

Assim,

$$
\begin{aligned}
\frac{1}{2}-\frac{1}{2^2}+\frac{1}{2^3}
&=\frac{1}{2}-\frac{1}{4}+\frac{1}{8}\\
&=\frac{4}{8}-\frac{2}{8}+\frac{1}{8}\\
&=\frac{4-2+1}{8}\\
&=\frac{3}{8}.
\end{aligned}
$$

Portanto, a alternativa correta é b).

#### Questão 2

O valor da expressão

$$
\left(\frac{5}{8}\cdot\frac{12}{25}-\frac{1}{10}\right)^{2}
$$

é:

a) $\frac{1}{20}$

b) $\frac{3}{20}$

c) $\frac{2}{15}$

d) $\frac{1}{25}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Realizamos inicialmente as operações dentro dos parênteses:

$$
\begin{aligned}
\frac{5}{8}\cdot\frac{12}{25}-\frac{1}{10}
&=\frac{5\cdot12}{8\cdot25}-\frac{1}{10}\\
&=\frac{3}{10}-\frac{1}{10}\\
&=\frac{2}{10}\\
&=\frac{1}{5}.
\end{aligned}
$$

Agora, elevamos o resultado ao quadrado:

$$\frac{1^2}{5^2}= \frac{1}{25}.$$

Portanto, a alternativa correta é d).

#### Questão 3

Analise as afirmações a seguir.

I. $\displaystyle \frac{1}{2}+\frac{1}{3}=\frac{5}{6}$.

II. $\displaystyle \frac{\frac{3}{4}}{\frac{9}{8}}=\frac{2}{3}$.

III. $\displaystyle \frac{5}{6}-\frac{7}{9}=\frac{1}{9}$.

Classificando cada afirmação como verdadeira (V) ou falsa (F), obtemos, respectivamente:

a) V, V, V.

b) V, V, F.

c) V, F, V.

d) F, V, F.

e) F, F, V.

##### Solução:

Analisamos cada afirmação separadamente.

Afirmação I

$$
\begin{aligned}
\frac{1}{2}+\frac{1}{3}
&=\frac{3}{6}+\frac{2}{6}\\
&=\frac{5}{6}.
\end{aligned}
$$

Logo, a afirmação I é verdadeira.

Afirmação II

$$
\begin{aligned}
\frac{\frac{3}{4}}{\frac{9}{8}}
&=\frac{3}{4}\cdot\frac{8}{9}\\
&=\frac{24}{36}\\
&=\frac{2}{3}.
\end{aligned}
$$

Logo, a afirmação II é verdadeira.

Afirmação III

$$
\begin{aligned}
\frac{5}{6}-\frac{7}{9}
&=\frac{15}{18}-\frac{14}{18}\\
&=\frac{1}{18}.
\end{aligned}
$$

Como

$$
\frac{1}{18}\neq\frac{1}{9},
$$

a afirmação III é falsa.

Portanto, a sequência correta é V, V, F.

A alternativa correta é b).

#### Questão 4

Considere as variáveis:

$$ a= 1 + \frac{1}{1+1}, \ \ b = 1 + \frac{1}{1+\frac{1}{1+1}}, \ \ c = 1 + \frac{1}{1+\frac{1}{1+\frac{1}{1+1}}}$$

O valor do produto $abc$ é dado por:

a) $4$

b) $3$

c) $\frac{3}{4}$

d) $\frac{2}{3}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Calculamos inicialmente o valor de $a$:

$$
\begin{aligned}
a
&=1+\frac{1}{1+1}\\
&=1+\frac{1}{2}\\
&=\frac{3}{2}.
\end{aligned}
$$

Calculamos agora o valor de $b$, começando pelas operações mais internas:

$$
\begin{aligned}
b
&=1+\frac{1}{1+\frac{1}{1+1}}\\
&=1+\frac{1}{1+\frac{1}{2}}\\
&=1+\frac{1}{\frac{3}{2}}\\
&=1+\frac{2}{3}\\
&=\frac{5}{3}.
\end{aligned}
$$

De modo semelhante, calculamos o valor de $c$

$$
\begin{aligned}
c
&=1+\frac{1}{1+\frac{1}{1+\frac{1}{1+1}}}\\
&=1+\frac{1}{1+\frac{1}{1+\frac{1}{2}}}\\
&=1+\frac{1}{1+\frac{1}{\frac{3}{2}}}\\
&=1+\frac{1}{1+\frac{2}{3}}\\
&=1+\frac{1}{\frac{5}{3}}\\
&=1+\frac{3}{5}\\
&=\frac{8}{5}.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
abc
&=\frac{3}{2}\cdot\frac{5}{3}\cdot\frac{8}{5}\\
&=\frac{3\cdot5\cdot8}{2\cdot3\cdot5}\\
&=\frac{8}{2}\\
&=4.
\end{aligned}
$$

Logo, a alternativa correta é a).
