<h1 style="color:black; font-size:22px;"> Aula 1: Propriedades Aritméticas Básicas </h1>

<hr style="border: 2px solid black;"> </div>

[🎥](https://youtu.be/jfY-dkn2_Lw)

Nesta aula, revisaremos algumas propriedades fundamentais dos números reais e a ordem das operações. Esses conhecimentos são úteis na resolução e na simplificação de expressões numéricas e algébricas, presentes em diversas áreas da Matemática e em disciplinas introdutórias de cursos das áreas de Ciências Exatas.

## 1.1 Propriedades 

#### <div style="color:#FF0000; font-weight:bold;">Comutatividade [🎥](https://youtu.be/jfY-dkn2_Lw7&t=73s) </div>



<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Para quaisquer $a$ e $b$ reais, temos:<br>
$$a + b = b + a \tag{1}$$<br>
$$a \cdot b = b \cdot a \tag {2}$$
</div>

#### Exemplo 1

A expressão $1 + 5$ pode ser calculada de duas formas:

$$1 + 5 = 5 + 1 = 6$$

Analogamente, a expressão $2 \cdot 3$ também pode ser calculada de duas formas:

$$2 \cdot 3 = 3 \cdot 2 = 6$$

##### Observação

Entre as quatro operações aritméticas básicas, apenas a adição e a multiplicação são comutativas. A subtração e a divisão, em geral, não são comutativas.

Por exemplo, na subtração:

$$
\begin{aligned}
7-5 &= 2, \\
5-7 &= -2.
\end{aligned}
$$

Portanto,

$$
7-5\neq5-7.
$$

Na divisão:

$$
\begin{aligned}
12:3 &= 4,\\
3:12 &= \frac{1}{4}.
\end{aligned}
$$

Logo,

$$
12:3\neq3:12.
$$

#### <div style="color:#FF0000; font-weight:bold;">Associatividade [🎥](https://youtu.be/jfY-dkn2_Lw7&t=236.5s)</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Para quaisquer $a$, $b$ e $c$ reais, temos:<br>
$$(a + b) + c = a + (b + c) \tag{3}$$<br>
$$(a \cdot b) \cdot c = a \cdot (b \cdot c) \tag {4}$$
</div>

#### Exemplo 2

Os termos da expressão 3+4+4 podem ser agrupados de duas formas:

i) $3 + 4 + 4 = (3 + 4) + 4 = 7 + 4 = 11$

ii) $3 + 4 + 4 = 3 + (4 + 4) = 3 + 8 = 11$

Analogamente, a expressão $3 \cdot 2 \cdot 4$ também pode ser calculada de duas formas:

i) $3 \cdot 2 \cdot 4 = (3 \cdot 2)\cdot 4 = 6 \cdot 4 = 24$

ii) $3 \cdot 2 \cdot 4 = 3 \cdot (2 \cdot 4) = 3 \cdot 8 = 24$

##### Observação

Assim como ocorre com a comutatividade, entre as quatro operações aritméticas básicas, apenas a adição e a multiplicação são associativas. A subtração e a divisão, em geral, não são associativas.

Por exemplo, na subtração:

$$
\begin{aligned}
7 - (5 - 2) & = 7 - 3 = 4, \\
(7 - 5) - 2 & = 2 - 2 = 0.
\end{aligned}
$$

Portanto,

$$
7 - (5 - 2) \neq (7 - 5) - 2.
$$

Na divisão:

$$
\begin{aligned}
12 : (4:2) &= 12:2 = 6,\\
(12:4):2 &= 3:2 = \frac{3}{2}.
\end{aligned}
$$

Logo,

$$
12:(4:2) \neq (12:4):2.
$$

#### <div style="color:#FF0000; font-weight:bold;">Distributividade [🎥](https://youtu.be/jfY-dkn2_Lw7&t=525s)</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Para quaisquer $a$, $b$ e $c$ reais, temos:<br>
$$a(b \pm c) = ab \pm ac \tag{5}$$<br>
$$(a \pm b) c = ac \pm bc \tag {6}$$
</div>

#### Exemplo 3

A expressão $2\cdot(3 + 1)$ pode ser calculada de duas formas:

i) $2 \cdot (3 + 1) = 2 \cdot 4 = 8$

ii) $2 \cdot (3 + 1) = 2\cdot 3 + 2 \cdot 1 = 6 + 2 = 8$

Analogamente, a expressão $(4-3)\cdot3$ também pode ser calculada de duas formas:

i) $(4-3)\cdot3 = 1 \cdot 3 = 3$

ii) $(4-3)\cdot3 = 4\cdot3 - 3\cdot3 =12 - 9 = 3$

## 1.2 Ordem das Operações [🎥](https://youtu.be/Ut92VX59I_Q)


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
A ordem dos cálculos em expressões que envolvem múltiplas operações é a seguinte:<br>
1. <strong>Delimitadores:</strong> parênteses $(\ )$, colchetes $[\ ]$ e chaves $\{\ \}$.<br>
Quando houver delimitadores aninhados, realizamos primeiro as operações contidas nos mais internos.<br>
2. <strong>Potenciações e radiciações:</strong> $a^n$ e $\sqrt[n]{a}$.<br>
3. <strong>Multiplicações e divisões:</strong> devem ser realizadas na ordem em que aparecem, da esquerda para a direita: $\cdot$ e $\div$.<br>
4. <strong>Adições e subtrações:</strong> devem ser realizadas na ordem em que aparecem, da esquerda para a direita: $+$ e $-$.
</div>

#### Exemplo 4 [🎥](https://youtu.be/Ut92VX59I_Q&t=164s)

Resolva a expressão numérica:

$$3\cdot(2 + 1) - 2\cdot(4-2)$$

##### Solução:

Temos que

$$
\begin{aligned}
3\cdot(2 + 1) - 2\cdot(4-2)
&= 3\cdot3 - 2\cdot2 \\
&= 9 - 4 \\
&= 5
\end{aligned}
$$


#### Exemplo 5 [🎥](https://youtu.be/QrB8IOg_STU)

Resolva a expressão numérica:

$$3\cdot2^{2} + 5 + \frac{3 + 5}{2}$$

##### Solução:

Temos que

$$
\begin{aligned}
3\cdot2^{2} + 5 + \frac{3 + 5}{2}
&= 3\cdot4 + 5 + \frac{8}{2} \\
&= 12 + 5 + 4 \\
&= 21
\end{aligned}
$$


#### Exemplo 6 [🎥](https://youtu.be/CTNlAaWiURQ)

Resolva a expressão numérica:

$$\frac{((20-8) - (6 - 2\cdot4)^{2})^{2}}{4 + 2\cdot6}$$

##### Solução:

Temos que

$$
\begin{aligned}
\frac{((20-8) - (6 - 2\cdot4)^{2})^{2}}{4 + 2\cdot6}
&= \frac{(12 - (6 - 8)^{2})^{2}}{4 + 12} \\
&= \frac{(12 - (-2)^{2})^{2}}{16} \\
&= \frac{(12 - 4)^{2}}{16} \\
&= \frac{8^{2}}{16} \\
&= \frac{64}{16} \\
&= 4
\end{aligned}
$$

#### Exemplo 7 [🎥](https://youtu.be/ae1AAXI0RaE)

Resolva a expressão numérica:

$$\sqrt{(15-9)^{2} + \left(6\sqrt{5 + 2\cdot 2} - 2\cdot5\right)^{2}}$$

##### Solução:

Temos que

$$
\begin{aligned}
\sqrt{(15-9)^{2} + \left(6\sqrt{5 + 2\cdot 2} - 2\cdot5\right)^{2}}
&= \sqrt{6^{2} + \left(6\sqrt{5 + 4} - 10\right)^{2}}\\
&= \sqrt{36 + \left(6\sqrt{9} - 10\right)^{2}} \\
&= \sqrt{36 + \left(6\cdot3 - 10\right)^{2}} \\
&= \sqrt{36 + \left(18 - 10\right)^{2}} \\
&= \sqrt{36 + 8^{2}} \\
&= \sqrt{36 + 64} \\
&= \sqrt{100} \\
&= 10
\end{aligned}
$$

Determine o valor das expressões numéricas:

a) $(2 + 3\cdot2)\cdot(7 - 2^{2})^{2}$

b) $\frac{(4\cdot2 + 5)^{2} - (4\cdot2 - 5)^{2}}{30 - 2\cdot(2 + 3)^{2}}$

c) $\left(\sqrt{3 + (7-3)^{2}-20:2} + 1\right)^{3}$

\end{aligned}

$$
\begin{aligned}
\frac{(4\cdot2 + 5)^{2} - (4\cdot2 - 5)^{2}}{30 - 2\cdot(2 + 3)^{2}}

\left(\sqrt{3 + (7-3)^{2}-20:2} + 1\right)^{3}
&= \left(\sqrt{3 + 4^{2}-10} + 1\right)^{3} \\
&= \left(\sqrt{3 + 16 -10} + 1\right)^{3} \\
&= \left(\sqrt{9} + 1\right)^{3} \\
&= (3 + 1)^{3} \\

\end{aligned}
$$

##### Solução do item d:

$$
\begin{aligned}
\frac{\left[3\cdot(5-2)+1\right]^2-\sqrt{144}}{2^3+3}

&=\frac{100-12}{11}\\

&=8.
\end{aligned}
$$

#### Exercício 2

Um estudante resolveu a expressão numérica da seguinte maneira:

$$
\begin{aligned}
18\div3\cdot2+4
&=18\div6+4\\
&=3+4\\
&=7.
\end{aligned}
$$

A resolução está correta? Justifique sua resposta e, caso seja necessário, apresente a resolução correta.

$$

18\div3\cdot2+4

As operações com frações e quocientes são fundamentais em diversas áreas da Matemática e aparecem com frequência nos cursos de Ciências Exatas.

## 2.1 Soma e Subtração

### <div style="color:#FF0000; font-weight:bold;">Frações com mesmo denominador</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> A soma ou a subtração de duas frações $\large \frac{a}{b}$ e $\large \frac{c}{b}$, com o mesmo denominador $b$ e $b \neq 0$, é dada por:<br>
$$\frac{a}{b} \pm \frac{c}{b} = \frac{a\pm c}{b}.$$<br>
Em outras palavras, mantemos o denominador e somamos ou subtraímos os numeradores.
</div>

#### Exemplo 1 [🎥](https://youtu.be/BNoLpZrU1js&t=203s)

Determine o valor das seguintes expressões:

a) $\frac{5}{6} + \frac{8}{6}$

b) $\frac{23}{9} - \frac{8}{9}$

c) $\frac{12}{5} + \frac{32}{10} - \frac{9}{15}$

##### Solução do item a:

Como as frações possuem o mesmo denominador, mantemos o denominador e somamos os numeradores:

$$
\begin{aligned}
\frac{5}{6}+\frac{8}{6}
&=\frac{5+8}{6}\\
&=\frac{13}{6}.
\end{aligned}
$$

##### Solução do item b:

Como as frações possuem o mesmo denominador, mantemos o denominador e subtraímos os numeradores:

$$
\begin{aligned}
\frac{23}{9}-\frac{8}{9}
&=\frac{23-8}{9}\\
&=\frac{15}{9}\\
&=\frac{5}{3}.
\end{aligned}
$$

##### Solução do item c:

Inicialmente, simplificamos as frações que ainda não estão escritas com denominador $5$:

$$
\frac{32}{10}=\frac{16}{5}
\ \ \text{e} \ \
\frac{9}{15}=\frac{3}{5}.
$$

Assim, obtemos frações com o mesmo denominador:

$$
\begin{aligned}
\frac{12}{5}+\frac{32}{10}-\frac{9}{15}
&=\frac{12}{5}+\frac{16}{5}-\frac{3}{5}\\
&=\frac{12+16-3}{5}\\
&=\frac{25}{5}\\
&=5.
\end{aligned}
$$

### <div style="color:#FF0000; font-weight:bold;">Frações com denominadores diferentes [🎥](https://youtu.be/8JKx5rt2Hww)</div> 

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> Para somar ou subtrair duas frações $\large \frac{a}{b}$ e $\large \frac{c}{d}$ com denominadores diferentes, devemos primeiro encontrar um denominador comum.<br>
Seja<br>
$$M=\operatorname{mmc}(b,d),$$<br>
em que $\operatorname{mmc}(b,d)$ representa o mínimo múltiplo comum de $b$ e $d$. Definimos<br><br>
$$x=\frac{M}{b} \qquad\text{e}\qquad y=\frac{M}{d}.$$<br>
Assim,<br>
$$\frac{a}{b} \pm \frac{c}{d} = \frac{ax\pm cy}{M},$$<br>
com $b\neq 0$ e $d\neq 0$.
</div>

#### Exemplo 2 [🎥](https://youtu.be/8JKx5rt2Hww&t=229s)

Determine o valor das seguintes expressões:

a) $\frac{5}{6} - \frac{7}{4}$

b) $\frac{5}{2} + \frac{2}{3} - \frac{1}{4}$

c) $\frac{9}{8} + \frac{1}{7} - 2$

##### Solução do item a:

Determinamos inicialmente o mínimo múltiplo comum dos denominadores:

$$
\operatorname{mmc}(6,4)=12.
$$

Como

$$
\frac{12}{6}=2
\qquad\text{e}\qquad
\frac{12}{4}=3,
$$

temos:

$$
\begin{aligned}
\frac{5}{6}-\frac{7}{4}
&=\frac{5\cdot2-7\cdot3}{12}\\
&=\frac{10-21}{12}\\
&=-\frac{11}{12}.
\end{aligned}
$$

##### Solução do item b:

Determinamos inicialmente o mínimo múltiplo comum dos denominadores:

$$
\operatorname{mmc}(2,3,4)=12.
$$

Como

$$
\frac{12}{2}=6, \ \ \frac{12}{3}=4 \ \ \text{e} \ \ \frac{12}{4}=3,
$$

temos:

$$
\begin{aligned}
\frac{5}{2}+\frac{2}{3}-\frac{1}{4}
&=\frac{5\cdot6+2\cdot4-1\cdot3}{12}\\
&=\frac{30+8-3}{12}\\
&=\frac{35}{12}.
\end{aligned}
$$

##### Solução do item c:

Podemos escrever o número inteiro $2$ na forma de uma fração:

$$
2=\frac{2}{1}.
$$

Determinamos, então, o mínimo múltiplo comum dos denominadores:

$$
\operatorname{mmc}(8,7,1)=56.
$$

Como

$$
\frac{56}{8}=7, \ \ \frac{56}{7}=8 \ \ \text{e} \ \ \frac{56}{1}=56,
$$

temos:

$$
\begin{aligned}
\frac{9}{8}+\frac{1}{7}-2
&=\frac{9}{8}+\frac{1}{7}-\frac{2}{1}\\
&=\frac{9\cdot7+1\cdot8-2\cdot56}{56}\\
&=\frac{63+8-112}{56}\\
&=-\frac{41}{56}.
\end{aligned}
$$

## 2.2 Produto e Divisão [🎥](https://youtu.be/J8yyVZjRj7o)

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> Para multiplicar duas frações, multiplicamos os numeradores entre si e os denominadores entre si.<br>
 Assim, para $\large \frac{a}{b}$ e $\large \frac{c}{d}$, com $b\neq 0$ e $d\neq 0$, temos:<br>
 $$\frac{a}{b}\cdot\frac{c}{d} = \frac{ac}{bd}.$$
</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> Para dividir por uma fração, multiplicamos pela sua fração inversa.<br>
 Assim, para $\large \frac{a}{b}$ e $\large \frac{c}{d}$, com $b\neq 0$, $c\neq 0$ e $d\neq 0$, temos:<br>
  $$\frac{\frac{a}{b}}{\frac{c}{d}}=\frac{ad}{bc}.$$
</div>

#### Exemplo 3 [🎥](https://youtu.be/J8yyVZjRj7o&t=199s)

Determine o valor das seguintes expressões:

a) $\frac{12}{5}\cdot\frac{28}{5}$

b) $\large \frac{\frac{21}{8}}{\frac{3}{5}}$

c) $\large \frac{\frac{6}{5} \cdot \frac{25}{12}}{\frac{15}{4}}$

##### Solução do item a:

Multiplicamos os numeradores entre si e os denominadores entre si:

$$
\begin{aligned}
\frac{12}{5}\cdot\frac{28}{5}
&=\frac{12\cdot28}{5\cdot5}\\
&=\frac{336}{25}.
\end{aligned}
$$

##### Solução do item b:

Para dividir por uma fração, multiplicamos pela sua fração inversa:

$$
\begin{aligned}
\frac{\frac{21}{8}}{\frac{3}{5}}
&=\frac{21}{8}\cdot\frac{5}{3}\\
&=\frac{21\cdot5}{8\cdot3}\\
&=\frac{7\cdot5}{8}\\
&=\frac{35}{8}.
\end{aligned}
$$

##### Solução do item c:

Para dividir por uma fração, multiplicamos pela sua inversa:

$$
\begin{aligned}
\frac{\frac{6}{5}\cdot\frac{25}{12}}{\frac{15}{4}}
&=\left(\frac{6}{5}\cdot\frac{25}{12}\right)\cdot\frac{4}{15}\\
&=\frac{6\cdot25\cdot4}{5\cdot12\cdot15}\\
&=\frac{1\cdot5\cdot2}{1\cdot1\cdot15}\\
&=\frac{10}{15}\\
&=\frac{2}{3}.
\end{aligned}
$$

#### Exemplo 4 [🎥](https://youtu.be/S-lrvN77s7s)

Resolva a expressão

$$\large \frac{1}{1 + \frac{1}{1 + \frac{1}{2}}}$$

##### Solução:

Iniciamos os cálculos pelas operações mais internas:

$$
\begin{aligned}
\frac{1}{1+\frac{1}{1+\frac{1}{2}}}
&=\frac{1}{1+\frac{1}{\frac{2+1}{2}}}\\
&=\frac{1}{1+\frac{1}{\frac{3}{2}}}\\
&=\frac{1}{1+\frac{2}{3}}\\
&=\frac{1}{\frac{5}{3}}\\
&=\frac{3}{5}.
\end{aligned}
$$

#### Exemplo 5 [🎥](https://youtu.be/NiO6i4nvURc)

Resolva a expressão

$$\sqrt{2 + \frac{16}{\large \frac{3 - \frac{1}{3}}{1 - \frac{2}{3}}}}.$$

##### Solução:

Iniciamos os cálculos pelas operações mais internas:

$$
\begin{aligned}
\sqrt{2+\frac{16}{\large\frac{3-\frac{1}{3}}{1-\frac{2}{3}}}}
&=
\sqrt{2+\frac{16}{\large\frac{\frac{9-1}{3}}{\frac{3-2}{3}}}}\\
&=
\sqrt{2+\frac{16}{\large\frac{\frac{8}{3}}{\frac{1}{3}}}}\\[4pt]
&=
\sqrt{2+\frac{16}{8}}\\[4pt]
&=
\sqrt{2+2}\\[4pt]
&=
\sqrt{4}\\[4pt]
&=2.
\end{aligned}
$$

Portanto, o valor da expressão é $2$.


## Exercícios

#### Exercício 1

Determine o valor das expressões:

a) $1 + \frac{1}{2} + \frac{1}{3}$

b) $\frac{3}{5}\cdot\frac{4}{9} - \frac{6}{5}\cdot\frac{10}{30}$

c) $\large \frac{\frac{1}{3}}{\frac{3 + \frac{4}{5}}{3}}$

d) $1 + \large \frac{12}{1 + \frac{6}{1+\frac{4}{1+1}}}$

##### Solução do item a:

Determinamos o mínimo múltiplo comum dos denominadores:

$$
\operatorname{mmc}(1,2,3)=6.
$$

Assim,

$$
\begin{aligned}
1+\frac{1}{2}+\frac{1}{3}
&=\frac{6}{6}+\frac{3}{6}+\frac{2}{6}\\[2pt]
&=\frac{6+3+2}{6}\\
&=\frac{11}{6}.
\end{aligned}
$$

##### Solução do item b:

Realizamos inicialmente as multiplicações:

$$
\begin{aligned}
\frac{3}{5}\cdot\frac{4}{9}
-\frac{6}{5}\cdot\frac{10}{30}
&=\frac{12}{45}-\frac{60}{150}\\
&=\frac{4}{15}-\frac{2}{5}\\
&=\frac{4}{15}-\frac{6}{15}\\
&=-\frac{2}{15}.
\end{aligned}
$$

##### Solução do item c:

Iniciamos os cálculos pelas operações mais internas:

$$
\begin{aligned}
\large \frac{\frac{1}{3}}{\frac{3+\frac{4}{5}}{3}}
&= \large \frac{\frac{1}{3}}{\frac{\frac{15}{5}+\frac{4}{5}}{3}}\\
&= \large \frac{\frac{1}{3}}{\frac{\frac{19}{5}}{3}}\\[2pt]
&= \large \frac{\frac{1}{3}}{\frac{19}{15}}\\[2pt]
&= \small \frac{1}{3}\cdot\frac{15}{19}\\[2pt]
&= \small \frac{5}{19}.
\end{aligned}
$$

##### Solução do item d:

Iniciamos os cálculos pelas operações mais internas:

$$
\begin{aligned}
1+\frac{12}{1+\frac{6}{1+\frac{4}{1+1}}}
&=1+\frac{12}{1+\frac{6}{1+\frac{4}{2}}}\\
&=1+\frac{12}{1+\frac{6}{1+2}}\\
&=1+\frac{12}{1+\frac{6}{3}}\\
&=1+\frac{12}{1+2}\\
&=1+\frac{12}{3}\\
&=1+4\\
&=5.
\end{aligned}
$$

Nesta aula, aprenderemos a reescrever e simplificar expressões algébricas utilizando propriedades já estudadas, como a comutatividade e a distributividade. Esses procedimentos são fundamentais para facilitar cálculos e preparar o estudo de equações, funções e outros conteúdos do Pré-Cálculo.

## 3.1 Expressões Algébricas Equivalentes

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> Duas expressões algébricas são equivalentes quando assumem o mesmo valor para todos os valores das variáveis para os quais ambas estão definidas. </div>

#### Exemplo 1 [🎥](https://youtu.be/KFDN75vXrOM&t=156s) 

Considere as expressões algébricas

$$2x+4y \ \ \text{e} \ \ 2(x+2y).$$

a) Calcule o valor de cada expressão para os seguintes pares de valores e compare os resultados obtidos:

$x=1$ e $y=2$;

$x=-1$ e $y=5$.

b) Calcule a diferença entre as duas expressões e mostre que ela é igual a zero para quaisquer valores de $x$ e $y$. Conclua a partir daí que as expressões são equivalentes.

##### Solução do item a:

Para $x=1$ e $y=2$, temos:

$$
2x+4y
=2\cdot1+4\cdot2
=2+8
=10
$$

e

$$
2(x+2y)
=2(1+2\cdot2)
=2(1+4)
=10.
$$

Para $x=-1$ e $y=5$, temos:

$$
2x+4y
=2\cdot(-1)+4\cdot5
=-2+20
=18
$$

e

$$
2(x+2y)
=2(-1+2\cdot5)
=2(-1+10)
=2\cdot9
=18
$$

Em ambos os pares de valores, as duas expressões apresentaram resultados iguais.

##### Solução do item b:

Calculando a diferença entre as expressões, obtemos:

$$
\begin{aligned}
(2x+4y)-2(x+2y)
&=2x+4y-(2x+4y)\\
&=2x+4y-2x-4y\\
&=0.
\end{aligned}
$$

Como a diferença entre as expressões é igual a zero para quaisquer valores de $x$ e $y$, temos:

$$
2x+4y=2(x+2y).
$$

Portanto, as expressões $2x+4y$ e $2(x+2y)$ são equivalentes.

#### Exemplo 2 [🎥](https://youtu.be/d_xrmV8-fGY) 

Considere as expressões algébricas

$$2(x+y) \ \ \text{e} \ \ 2x+y.$$

a) Calcule o valor de cada expressão para $x=1$ e $y=0$ e compare os resultados obtidos.

b) O resultado do item anterior é suficiente para concluir que as expressões são equivalentes? Justifique.

##### Solução do item a:

Para $x=1$ e $y=0$, temos:

$$
2(x+y)
=2(1+0)
=2
$$

e

$$
2x+y
=2\cdot1+0
=2.
$$

Portanto, para $x=1$ e $y=0$, as duas expressões apresentam o mesmo resultado.

##### Solução do item b:

O resultado obtido no item anterior não é suficiente para concluir que as expressões são equivalentes, pois a igualdade deveria ser válida para todos os pares de valores de $x$ e $y$.

Consideremos, por exemplo, $x=1$ e $y=2$. Para a primeira expressão, temos:

$$
2(x+y)
=2(1+2)
=6.
$$

Para a segunda expressão, obtemos:

$$
2x+y
=2\cdot1+2
=4.
$$

Assim,

$$
2(x+y)\neq 2x+y
$$

quando $x=1$ e $y=2$.

Portanto, as expressões não são equivalentes. O par $x=1$ e $y=2$ é um contraexemplo para a afirmação de que

$$
2(x+y)=2x+y.
$$

Um único contraexemplo é suficiente para mostrar que uma igualdade não é válida para todos os valores das variáveis.

## 3.2 Manipulação e Simplificação de Expressões Algébricas [🎥](https://youtu.be/M8PejPYHu8g) 

Simplificar uma expressão algébrica significa obter uma forma equivalente mais compacta ou mais conveniente. Para isso, podemos aplicar propriedades algébricas, eliminar parênteses, reunir termos semelhantes ou colocar fatores comuns em evidência.

#### Exemplo 3 [🎥](https://youtu.be/M8PejPYHu8g&t=120s)

Reescreva as expressões a seguir em outras formas equivalentes:

a) $2(x+y)$

b) $2x + 3y + 2a + 3b$

##### Solução do item a:

Utilizando as propriedades distributiva e comutativa, temos:

$$\begin{aligned}
2(x+y)
&=2x + 2y\\[2pt]
2(x+y)
&=2(y+x)\\[2pt]
2(x+y)
&=2x+2y=2y+2x
\end{aligned}$$

Existem ainda outras formas equivalentes.

##### Solução do item b:

Utilizando a propriedade comutativa e colocando fatores comuns em evidência, temos:

$$\begin{aligned}
2x+3y+2a+3b
&=2x+2a+3y+3b\\
&=2(x+a)+3(y+b).
\end{aligned}$$

Também podemos escrever:

$$\begin{aligned}
2x+3y+2a+3b
&=3y+3b+2x+2a\\
&=3(y+b)+2(x+a).
\end{aligned}$$

Existem ainda outras formas equivalentes.

#### <div style="color:#FF0000; font-weight:bold;">Termos Semelhantes [🎥](https://youtu.be/H8f9PeF67YU)</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;"> Termos semelhantes são aqueles que possuem a mesma parte literal, com as mesmas variáveis elevadas aos mesmos expoentes.</div>

#### Exemplo 4 Termos Semelhantes [🎥](https://youtu.be/H8f9PeF67YU&t=33s)

Simplifique a expressão:

$$
2(3m+x)+5x+2m.
$$

##### Solução

Primeiramente, aplicamos a propriedade distributiva. Depois, reunimos os termos semelhantes:

$$
\begin{aligned}
2(3m+x)+5x+2m
&=6m+2x+5x+2m\\
&=6m+2m+2x+5x\\
&=8m+7x.
\end{aligned}
$$

Neste exemplo, a forma simplificada possui menos operações do que a expressão original.

#### Exemplo 5 [🎥](https://youtu.be/mP28onFBG9k)

Simplifique a expressão: 

$$xy + zw - xw - yz$$

##### Solução

Agrupando convenientemente os termos, temos:

$$
\begin{aligned}
xy+zw-xw-yz
&=xy-xw+zw-yz\\
&=x(y-w)+z(w-y).
\end{aligned}
$$

Como $w-y=-(y-w)$, segue que:

$$
\begin{aligned}
x(y-w)+z(w-y)
&=x(y-w)-z(y-w)\\
&=(x-z)(y-w).
\end{aligned}
$$

Portanto,

$$
xy+zw-xw-yz=(x-z)(y-w).
$$

A expressão original possui quatro multiplicações, uma adição e duas subtrações, enquanto a nova forma possui uma multiplicação e duas subtrações. Portanto, obtivemos uma forma equivalente com menos operações.

#### Exemplo 6 [🎥](https://youtu.be/iewrGt6-3l4)

Simplifique a expressão:

$$\frac{xa + yb + xb + ya}{x^{2} + xy}$$

##### Solução

Inicialmente, devemos observar que:

$$
x^2+xy=x(x+y).
$$

Portanto, a expressão está definida apenas quando

$$
x\neq 0
\ \ \text{e} \ \
x+y\neq 0.
$$

Reorganizando e fatorando o numerador, temos:

$$
\begin{aligned}
\frac{xa+yb+xb+ya}{x^2+xy}
&=\frac{xa+ya+xb+yb}{x(x+y)}\\
&=\frac{a(x+y)+b(x+y)}{x(x+y)}\\
&=\frac{(x+y)(a+b)}{x(x+y)}\\
&=\frac{a+b}{x}.
\end{aligned}
$$

Assim,

$$\frac{xa+yb+xb+ya}{x^2+xy} = \frac{a+b}{x},$$

desde que

$$
x\neq 0
\ \ \text{e} \ \
x\neq -y.
$$

Calcule o valor de cada expressão para os seguintes pares de valores e compare os resultados obtidos:

$x=2$ e $y=1$;

As expressões são equivalentes? Justifique sua resposta.

3x+6y

3x+6y

e

=3(-1+2\cdot4)
=3(-1+8)

=21.
$$

Aplicando a propriedade distributiva à segunda expressão, obtemos:

$$
3(x+2y)
=3x+3\cdot2y

Portanto, as expressões

$$
3x+6y
\ \ \text{e} \ \

#### Exercício 2

Considere as expressões algébricas

$$4x(x+y) \ \ \text{e} \ \ 4y(x+y).$$

As expressões são equivalentes? Justifique sua resposta.

##### Solução:

Para que as expressões fossem equivalentes, elas deveriam assumir o mesmo valor para todos os valores de $x$ e $y$.

Consideremos, por exemplo, $x=1$ e $y=0$. Para a primeira expressão, temos:

$$
4x(x+y)
=4\cdot1(1+0)
=4.
$$

Para a segunda expressão, obtemos:

$$
4y(x+y)
=4\cdot0(1+0)
=0.
$$

Assim,

$$
4x(x+y)\neq 4y(x+y)
$$

quando $x=1$ e $y=0$. Portanto, as expressões não são equivalentes.

O par $x=1$ e $y=0$ é um contraexemplo para a afirmação de que

$$
4x(x+y)=4y(x+y).
$$

#### Exercício 3

Simplifique as expressões algébricas a seguir:

a) $2(3a-b)+3b-6a$

b) $5am+5an-2bm-2bn$

Aplicando a propriedade distributiva e reunindo os termos semelhantes, temos:

$$
\begin{aligned}
2(3a-b)+3b-6a
&=6a-2b+3b-6a\\
&=6a-6a-2b+3b\\

\end{aligned}

Portanto,

$$
2(3a-b)+3b-6a=b.

$$
\begin{aligned}
5am+5an-2bm-2bn
&=5a(m+n)-2b(m+n)\\
&=(5a-2b)(m+n).
\end{aligned}
$$

Portanto,

$$
5am+5an-2bm-2bn=(5a-2b)(m+n).
$$

#### Exercício 4

Simplifique a expressão algébrica a seguir e indique as condições necessárias para que a expressão original esteja definida:

$$
\frac{(2a+b)x+(2a+b)y}{(2a-b)x+(2a-b)y}.
$$

##### Solução:

Colocando os fatores comuns em evidência no numerador e no denominador, temos:

$$
\begin{aligned}
\frac{(2a+b)x+(2a+b)y}{(2a-b)x+(2a-b)y}
&=\frac{(2a+b)(x+y)}{(2a-b)(x+y)}\\
&=\frac{2a+b}{2a-b}.
\end{aligned}
$$

Para que a expressão original esteja definida, seu denominador deve ser diferente de zero. Portanto,

$$
(2a-b)(x+y)\neq 0.
$$

Assim, devemos ter:

$$
2a-b\neq 0
\ \ \text{e} \\
x+y\neq 0.
$$

Equivalentemente,

$$
b\neq 2a
\ \ \text{e} \\
x\neq -y.
$$

Portanto,

$$
\frac{(2a+b)x+(2a+b)y}{(2a-b)x+(2a-b)y} = \frac{2a+b}{2a-b},
$$

desde que

$$
b\neq 2a
\ \ \text{e} \\
x\neq -y.
$$

## 4.1 Fórmulas Notáveis (Grau 2) [🎥](https://youtu.be/iewrGt6-3l4)

O processo de simplificação e manipulação de algumas expressões algébricas requer o uso de fórmulas especiais. Vejamos três delas:

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
    
1. $(a + b)^{2} = a^{2} + 2ab + b^{2}$
    
2. $(a - b)^{2} = a^{2} - 2ab + b^{2}$
    
3. $(a+b)(a-b) = a^{2} - b^{2}$
    
</div>


As duas primeiras fórmulas são chamadas de quadrado da soma e quadrado da diferença, respectivamente. A terceira é conhecida como diferença de dois quadrados.

#### Exemplo 1  [🎥](https://youtu.be/OkI1Km41Jss) 

Expanda cada uma das expressões a seguir:

a) $(2+3)^{2}$

b) $(x-y)^{2}$

c) $(2x+3y)^{2}$

d) $(3+x)(3-x)$

##### Solução do item a:

Usaremos a fórmula $(a+b)^{2}=a^{2}+2ab+b^{2}$ com $a=2$ e $b=3$:

$$
\begin{aligned}
(2+3)^{2}
&=2^{2}+2\cdot2\cdot3+3^{2}\\
&=4+12+9\\
&=25.
\end{aligned}
$$

##### Solução do item b:

Usaremos a fórmula $(a-b)^{2}=a^{2}-2ab+b^{2}$ com $a=x$ e $b=y$:

$$
\begin{aligned}
(x-y)^{2}
&=x^{2}-2xy+y^{2}.
\end{aligned}
$$

##### Solução do item c:

Usaremos a fórmula $(a+b)^{2}=a^{2}+2ab+b^{2}$ com $a=2x$ e $b=3y$:

$$
\begin{aligned}
(2x+3y)^{2}
&=(2x)^{2}+2(2x)(3y)+(3y)^{2}\\
&=4x^{2}+12xy+9y^{2}.
\end{aligned}
$$

##### Solução do item d:

Usaremos a fórmula $(a+b)(a-b)=a^{2}-b^{2}$ com $a=3$ e $b=x$:

$$
\begin{aligned}
(3+x)(3-x)
&=3^{2}-x^{2}\\
&=9-x^{2}.
\end{aligned}
$$

c) $\displaystyle 25+4x^{2}-20x$

$$

$$

##### Solução do item b:

Note que

$$
\begin{aligned}
t^{2}-16
&=t^{2}-4^{2}\\

$$

##### Solução do item c:

Note que $25+4x^{2}-20x$ pode ser reorganizada como $4x^{2}-20x+25$.

Assim,

$$

$$

#### Exemplo 3

Simplifique a expressão racional:

$$\frac{x^{2}-6x+9}{x^{2}-9}$$

##### Solução:

Inicialmente, fatoramos o numerador e o denominador:

$$
x^{2}-6x+9
=x^{2}-2\cdot x\cdot3+3^{2}
=(x-3)^{2},
$$

e

$$
x^{2}-9
=x^{2}-3^{2}
=(x+3)(x-3).
$$

Portanto,

$$
\begin{aligned}
\frac{x^{2}-6x+9}{x^{2}-9}
&=\frac{(x-3)^{2}}{(x+3)(x-3)}\\
&=\frac{(x-3)(x-3)}{(x+3)(x-3)}\\
&=\frac{x-3}{x+3}.
\end{aligned}
$$

Entretanto, o denominador da expressão original não pode ser igual a zero. Como

$$
x^{2}-9=(x-3)(x+3),
$$

devemos ter

$$
x\neq 3
\ \ \text{e} \ \
x\neq -3.
$$

Assim,

$$
\frac{x^{2}-6x+9}{x^{2}-9}
=\frac{x-3}{x+3},
\ \ x\neq -3 \ \text{ e } \ x\neq 3.
$$

## 4.2 Fatoração de Trinômios do Segundo Grau

As fórmulas notáveis permitem reconhecer e fatorar rapidamente expressões como

$$
x^{2}+4x+4
\qquad\text{e}\qquad
x^{2}-9.
$$

Entretanto, nem todo trinômio do segundo grau pode ser identificado diretamente como um quadrado perfeito ou como uma diferença de dois quadrados. Por exemplo, considere a expressão

$$
x^{2}-5x+6.
$$

Existe uma fatoração para essa expressão?

A resposta é sim. Quando um trinômio do segundo grau não corresponde diretamente a uma fórmula notável, podemos procurar as raízes da equação associada e utilizá-las para obter sua fatoração.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

Considere o polinômio do segundo grau

$$
p(x)=ax^{2}+bx+c,
\qquad a\neq0.
$$

As raízes $x_{1}$ e $x_{2}$ da equação associada

$$
ax^{2}+bx+c=0
$$

são dadas por

$$
x_{1}=\frac{-b+\sqrt{\Delta}}{2a}
\qquad\text{e}\qquad
x_{2}=\frac{-b-\sqrt{\Delta}}{2a},
$$

em que

$$
\Delta=b^{2}-4ac.
$$

Conhecidas as raízes, o polinômio pode ser fatorado na forma

$$
p(x)=a(x-x_{1})(x-x_{2}).
$$

Portanto,

$$
ax^{2}+bx+c=a(x-x_{1})(x-x_{2}).
$$

</div>


Quando os coeficientes $a$, $b$ e $c$ são reais:

* se $\Delta>0$, o polinômio possui duas raízes reais distintas;
* se $\Delta=0$, o polinômio possui uma raiz real dupla;
* se $\Delta<0$, o polinômio possui duas raízes complexas não reais.

#### Exemplo 4

Fatore a expressão a seguir:

$$\displaystyle x^{2}-5x+6.$$

##### Solução:

Para o polinômio

$$
x^{2}-5x+6,
$$

temos

$$
a=1,
\qquad b=-5,
\qquad c=6.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^{2}-4ac\\
&=(-5)^{2}-4\cdot1\cdot6\\
&=25-24\\
&=1.
\end{aligned}
$$

As raízes são

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{-(-5)\pm\sqrt{1}}{2\cdot1}\\
&=\frac{5\pm1}{2}.
\end{aligned}
$$

Portanto,

$$
x_{1}=3
\qquad\text{e}\qquad
x_{2}=2.
$$

Como $a=1$, temos

$$
\begin{aligned}
x^{2}-5x+6
&=(x-x_{1})(x-x_{2})\\
&=(x-3)(x-2).
\end{aligned}
$$

Logo,

$$
x^{2}-5x+6=(x-2)(x-3).
$$

#### Exemplo 5

Simplifique a expressão a seguir e indique os valores para os quais a expressão original não está definida:

$$
\displaystyle \frac{2x^{2}+x-6}{x^{2}-x-6}.
$$

##### Solução:

Inicialmente, fatoramos o numerador. Para a equação

$$
2x^{2}+x-6=0,
$$

temos

$$
a=2,\qquad b=1,\qquad c=-6.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^{2}-4ac\\
&=1^{2}-4\cdot2\cdot(-6)\\
&=1+48\\
&=49.
\end{aligned}
$$

As raízes são

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{-1\pm\sqrt{49}}{2\cdot2}\\
&=\frac{-1\pm7}{4}.
\end{aligned}
$$

Portanto,

$$
x_{1}=\frac{3}{2}
\qquad\text{e}\qquad
x_{2}=-2.
$$

Assim,

$$
\begin{aligned}
2x^{2}+x-6
&=2\left(x-\frac{3}{2}\right)(x+2)\\
&=(2x-3)(x+2).
\end{aligned}
$$

Agora, fatoramos o denominador. Para a equação

$$
x^{2}-x-6=0,
$$

temos

$$
a=1,\qquad b=-1,\qquad c=-6.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^{2}-4ac\\
&=(-1)^{2}-4\cdot1\cdot(-6)\\
&=1+24\\
&=25.
\end{aligned}
$$

As raízes são

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{-(-1)\pm\sqrt{25}}{2\cdot1}\\
&=\frac{1\pm5}{2}.
\end{aligned}
$$

Portanto,

$$
x_{1}=3
\qquad\text{e}\qquad
x_{2}=-2.
$$

Logo,

$$
x^{2}-x-6=(x-3)(x+2).
$$

Substituindo as fatorações na expressão, obtemos

$$
\begin{aligned}
\frac{2x^{2}+x-6}{x^{2}-x-6}
&=\frac{(2x-3)(x+2)}{(x-3)(x+2)}\\
&=\frac{2x-3}{x-3}.
\end{aligned}
$$

A expressão original não está definida quando

$$
(x-3)(x+2)=0,
$$

isto é, quando

$$
x=3
\qquad\text{ou}\qquad
x=-2.
$$

Portanto,

$$
\displaystyle \frac{2x^{2}+x-6}{x^{2}-x-6} = \frac{2x-3}{x-3},
$$

desde que $x\neq3$ e $x\neq-2$.

#### Observação:

Alguns polinômios do segundo grau incompletos podem ser fatorados sem a aplicação da fórmula de Bhaskara.

##### Caso 1: $b = 0 \Rightarrow ax^{2} + c$

Considere o polinômio

$$
2x^{2}-18.
$$

###### Solução:

Nesse caso,

$$
a=2,
\qquad b=0,
\qquad c=-18.
$$

Podemos resolver a equação correspondente:

$$
\begin{aligned}
2x^{2}-18&=0\\
2x^{2}&=18\\
x^{2}&=9\\
x&=\pm3.
\end{aligned}
$$

Portanto, as raízes são

$$
x_{1}=-3
\qquad\text{e}\qquad
x_{2}=3.
$$

Como o coeficiente do termo $x^{2}$ é $a=2$, temos

$$
\begin{aligned}
2x^{2}-18
&=2(x-x_{1})(x-x_{2})\\
&=2(x+3)(x-3).
\end{aligned}
$$

Também poderíamos obter essa fatoração colocando o fator comum $2$ em evidência e utilizando a diferença de dois quadrados:

$$
\begin{aligned}
2x^{2}-18
&=2(x^{2}-9)\\
&=2(x^{2}-3^{2})\\
&=2(x+3)(x-3).
\end{aligned}
$$

##### Caso 2: $c = 0 \Rightarrow ax^{2} + bx$

Considere o polinômio

$$
x^{2}-3x.
$$

###### Solução:


Nesse caso,

$$
a=1,
\qquad b=-3,
\qquad c=0.
$$

Podemos colocar $x$ em evidência:

$$
\begin{aligned}
x^{2}-3x
&=x(x-3)\\
&=(x-0)(x-3).
\end{aligned}
$$

Assim, as raízes de

$$
x^{2}-3x=0
$$

são

$$
x_{1}=0
\qquad\text{e}\qquad
x_{2}=3.
$$

De maneira geral, quando $c=0$, temos

$$
ax^{2}+bx=x(ax+b).
$$

## 5.1 Fórmulas Notáveis (Grau 3)

Algumas fórmulas de grau $3$ são especialmente importantes em disciplinas como Cálculo:

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

1. $(a + b)^{3} = a^{3} + 3a^{2}b + 3ab^{2} + b^{3}$<br><br>
2. $(a - b)^{3} = a^{3} - 3a^{2}b + 3ab^{2} - b^{3}$<br><br>
3. $(a + b)(a^{2} - ab + b^{2}) = a^{3} + b^{3}$<br><br>
4. $(a - b)(a^{2} + ab + b^{2}) = a^{3} - b^{3}$

</div>

#### Exemplo 1

Expanda cada uma das expressões a seguir:

a) $(x + y)^{3}$

b) $(x - 3)^{3}$

c) $(x + 1)(x^{2} - x + 1)$

##### Solução do item a:

A expressão $(x+y)^{3}$ tem a forma apresentada no item 1. Portanto,

$$
\begin{aligned}
(x+y)^{3}
&=x^{3}+3x^{2}y+3xy^{2}+y^{3}.
\end{aligned}
$$

##### Solução do item b:

A expressão $(x-3)^{3}$ tem a forma apresentada no item 2. Nesse caso, temos $a=x$ e $b=3$. Assim,

$$
\begin{aligned}
(x-3)^{3}
&=x^{3}-3x^{2}\cdot3+3x\cdot3^{2}-3^{3}\\
&=x^{3}-9x^{2}+27x-27.
\end{aligned}
$$

##### Solução do item c:

A expressão possui a forma correspondente à fatoração da soma de dois cubos, apresentada no item 3:

$$
a^{3}+b^{3}=(a+b)(a^{2}-ab+b^{2}).
$$

Tomando $a=x$ e $b=1$, obtemos:

$$
\begin{aligned}
(x+1)(x^{2}-x+1)
&=(x+1)(x^{2}-x\cdot1+1^{2})\\
&=x^{3}+1^{3}\\
&=x^{3}+1.
\end{aligned}
$$

<hr style="border: 2px solid black;"> </div>

A potenciação e a radiciação aparecem com frequência na simplificação de expressões numéricas e algébricas. Nesta aula, revisaremos as principais propriedades dessas operações e veremos como utilizá-las diretamente na resolução de alguns exemplos.

## 6.1 Potenciação

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Dados um número real $a$ e um número inteiro positivo $n$, a potência $a^n$ representa o produto de $n$ fatores iguais a $a$.
$$
a^{n} = \underbrace{a\cdot a\cdot a\cdots a}_{n\text{ fatores}}.
$$
    
Na expressão $a^{n}$, o número $a$ é chamado de base, enquanto $n$ é chamado de expoente.    
</div>

As principais propriedades da potenciação são apresentadas a seguir.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

Para $a,b\in\mathbb{R}$ e $m,n\in\mathbb{Z}$, sempre que as expressões estiverem definidas, temos:<br><br>

1. $a^{0}=1$, com $a\neq0$<br><br>
2. $a^{m}\cdot a^{n}=a^{m+n}$<br><br>
3. $\displaystyle\frac{a^{m}}{a^{n}}=a^{m-n}$, com $a\neq0$<br><br>
4. $\displaystyle a^{-n}=\frac{1}{a^{n}}$, com $a\neq0$<br><br>
5. $\left(a^{m}\right)^{n}=a^{mn}$<br><br>
6. $(ab)^{n}=a^{n}b^{n}$<br><br>
7. $\displaystyle\left(\frac{a}{b}\right)^{n}=\frac{a^{n}}{b^{n}}$, com $b\neq0$

</div>

#### Exemplo 1

Use a definição de potenciação e as propriedades apresentadas acima para calcular as seguintes potências e expressões:

a) $2^{5}$

b) $7^{0}$

c) $2^{3}\cdot2^{5}$

d) $\displaystyle\frac{3^{10}}{3^{8}}$

e) $2^{-4}$

f) $\left(2^{3}\right)^{2}$

g) $(2\cdot3)^{3}$

h) $\displaystyle\frac{6^{3}}{3^{3}}$

##### Solução:

a) Pela definição de potenciação,

$$
\begin{aligned}
2^{5}
&=2\cdot2\cdot2\cdot2\cdot2\\
&=32.
\end{aligned}
$$

b) Como a base é diferente de zero, aplicamos a propriedade do expoente zero:

$$
7^{0}=1.
$$

c) Como as potências possuem a mesma base, adicionamos os expoentes:

$$
\begin{aligned}
2^{3}\cdot2^{5}
&=2^{3+5}\\
&=2^{8}\\
&=256.
\end{aligned}
$$

Essa propriedade também pode ser compreendida pela definição de potenciação:

$$
\begin{aligned}
2^{3}\cdot2^{5}
&=(2\cdot2\cdot2)(2\cdot2\cdot2\cdot2\cdot2)\\
&=8\cdot32\\
&=256.
\end{aligned}
$$

d) Como as potências possuem a mesma base, subtraímos os expoentes:

$$
\begin{aligned}
\frac{3^{10}}{3^{8}}
&=3^{10-8}\\
&=3^{2}\\
&=9.
\end{aligned}
$$

Note que calcular separadamente $3^{10}$ e $3^{8}$ para, em seguida, dividir os resultados seria bastante trabalhoso e desnecessário.

e) Aplicando a propriedade do expoente negativo, obtemos:

$$
\begin{aligned}
2^{-4}
&=\frac{1}{2^{4}}\\
&=\frac{1}{16}.
\end{aligned}
$$

f) Aplicando a propriedade da potência de uma potência, multiplicamos os expoentes:

$$
\begin{aligned}
\left(2^{3}\right)^{2}
&=2^{3\cdot2}\\
&=2^{6}\\
&=64.
\end{aligned}
$$

Também podemos calcular inicialmente a potência que está entre parênteses:

$$
\begin{aligned}
\left(2^{3}\right)^{2}
&=8^{2}\\
&=64.
\end{aligned}
$$

g) Podemos calcular inicialmente o produto:

$$
\begin{aligned}
(2\cdot3)^{3}
&=6^{3}\\
&=216.
\end{aligned}
$$

Também podemos aplicar a propriedade da potência de um produto:

$$
\begin{aligned}
(2\cdot3)^{3}
&=2^{3}\cdot3^{3}\\
&=8\cdot27\\
&=216.
\end{aligned}
$$

h) Como o numerador e o denominador possuem o mesmo expoente, aplicamos a propriedade da potência de um quociente no sentido inverso:

$$
\begin{aligned}
\frac{6^{3}}{3^{3}}
&=\left(\frac{6}{3}\right)^{3}\\
&=2^{3}\\
&=8.
\end{aligned}
$$

Nesse caso, a aplicação da propriedade permite simplificar o quociente antes de calcular a potência, tornando a resolução mais direta.

#### Exemplo 2

Simplifique as expressões e calcule os seus valores:

a) $\displaystyle \frac{2^{6}\cdot2^{8}}{2^{12}}$

b) $\displaystyle \frac{2^{7}-2^{5}}{2^{6}}$

c) $\displaystyle \frac{2^{16}+2^{19}}{2^{16}-2^{14}}$

##### Solução do item a:

Inicialmente, aplicamos a propriedade do produto de potências de mesma base. Em seguida, utilizamos a propriedade do quociente:

$$
\begin{aligned}
\frac{2^{6}\cdot2^{8}}{2^{12}}
&=\frac{2^{6+8}}{2^{12}}\\
&=\frac{2^{14}}{2^{12}}\\
&=2^{14-12}\\
&=2^{2}\\
&=4.
\end{aligned}
$$

##### Solução do item b:

Como o numerador contém uma subtração, não podemos simplesmente subtrair os expoentes. Devemos colocar a potência de menor expoente em evidência:

$$
\begin{aligned}
2^{7}-2^{5}
&=2^{5}\cdot2^{2}-2^{5}\\
&=2^{5}\left(2^{2}-1\right)\\
&=2^{5}(4-1)\\
&=3\cdot2^{5}.
\end{aligned}
$$

Assim,

$$
\begin{aligned}
\frac{2^{7}-2^{5}}{2^{6}}
&=\frac{3\cdot2^{5}}{2^{6}}\\
&=3\cdot2^{5-6}\\
&=3\cdot2^{-1}\\
&=\frac{3}{2}.
\end{aligned}
$$

##### Solução do item c:

No numerador, colocamos $2^{16}$ em evidência:

$$
\begin{aligned}
2^{16}+2^{19}
&=2^{16}+2^{16}\cdot2^{3}\\
&=2^{16}\left(1+2^{3}\right)\\
&=2^{16}(1+8)\\
&=9\cdot2^{16}.
\end{aligned}
$$

No denominador, colocamos $2^{14}$ em evidência:

$$
\begin{aligned}
2^{16}-2^{14}
&=2^{14}\cdot2^{2}-2^{14}\\
&=2^{14}\left(2^{2}-1\right)\\
&=2^{14}(4-1)\\
&=3\cdot2^{14}.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
\frac{2^{16}+2^{19}}{2^{16}-2^{14}}
&=\frac{9\cdot2^{16}}{3\cdot2^{14}}\\
&=\frac{9}{3}\cdot2^{16-14}\\
&=3\cdot2^{2}\\
&=3\cdot4\\
&=12.
\end{aligned}
$$

#### Exemplo 3

Simplifique a seguinte expressão:

$$
\frac{3^{x+2}-3^{x}}{3^{x+1}+3^{x}}.
$$

##### Solução:

No numerador, podemos colocar $3^{x}$ em evidência:

$$
\begin{aligned}
3^{x+2}-3^{x}
&=3^{x}\cdot3^{2}-3^{x}\\
&=3^{x}\left(3^{2}-1\right)\\
&=3^{x}(9-1)\\
&=8\cdot3^{x}.
\end{aligned}
$$


No denominador, também colocamos $3^{x}$ em evidência:

$$
\begin{aligned}
3^{x+1}+3^{x}
&=3^{x}\cdot3+3^{x}\\
&=3^{x}(3+1)\\
&=4\cdot3^{x}.
\end{aligned}
$$

Assim,

$$
\begin{aligned}
\frac{3^{x+2}-3^{x}}{3^{x+1}+3^{x}}
&=\frac{8\cdot3^{x}}{4\cdot3^{x}}\\
&=\frac{8}{4}\\
&=2.
\end{aligned}
$$

Como $3^{x}\neq0$ para todo $x\in\mathbb{R}$, o cancelamento é válido.

#### Exemplo 4

Calcule o valor da expressão a seguir para $x=2$ e $a=10$:

$$
\frac{x^{a+3}-x^{a}}{x^{a+2}-x^{a}}.
$$

##### Solução:

Substituir imediatamente $x=2$ e $a=10$ exigiria o cálculo de potências relativamente grandes. Antes da substituição, podemos simplificar a expressão colocando $x^{a}$ em evidência no numerador e no denominador:

$$
\begin{aligned}
\frac{x^{a+3}-x^{a}}{x^{a+2}-x^{a}}
&=
\frac{x^{a}\cdot x^{3}-x^{a}}
{x^{a}\cdot x^{2}-x^{a}}\\
&=
\frac{x^{a}(x^{3}-1)}
{x^{a}(x^{2}-1)}\\
&=
\frac{x^{3}-1}{x^{2}-1}.
\end{aligned}
$$

Como $x=2$, temos $x^{a}=2^{10}\neq0$, de modo que o cancelamento é válido. Substituindo $x=2$ na expressão simplificada, obtemos:

$$
\begin{aligned}
\frac{x^{3}-1}{x^{2}-1}
&=
\frac{2^{3}-1}{2^{2}-1}\\
&=
\frac{8-1}{4-1}\\
&=
\frac{7}{3}.
\end{aligned}
$$

#### Exemplo 5

Calcule o valor da expressão

$$
\frac{a^{4}b^{2}+2a^{3}b^{3}+a^{2}b^{4}}
{a^{3}b^{2}+a^{2}b^{3}},
$$

para $a=7$ e $b=3$.

##### Solução:

Antes de substituir os valores de $a$ e $b$, podemos simplificar a expressão.

No numerador, colocamos $a^{2}b^{2}$ em evidência:

$$
\begin{aligned}
a^{4}b^{2}+2a^{3}b^{3}+a^{2}b^{4}
&=a^{2}b^{2}\left(a^{2}+2ab+b^{2}\right)\\
&=a^{2}b^{2}(a+b)^{2}.
\end{aligned}
$$

No denominador, também colocamos $a^{2}b^{2}$ em evidência:

$$
\begin{aligned}
a^{3}b^{2}+a^{2}b^{3}
&=a^{2}b^{2}(a+b).
\end{aligned}
$$

Assim,

$$
\begin{aligned}
\frac{a^{4}b^{2}+2a^{3}b^{3}+a^{2}b^{4}}
{a^{3}b^{2}+a^{2}b^{3}}
&=
\frac{a^{2}b^{2}(a+b)^{2}}
{a^{2}b^{2}(a+b)}\\
&=a+b.
\end{aligned}
$$

Substituindo $a=7$ e $b=3$, obtemos:

$$
\begin{aligned}
a+b
&=7+3\\
&=10.
\end{aligned}
$$

## 6.2 Radiciação

### 6.2.1 Propriedades Básicas

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

Dados um número real $a$ e um número natural $n\geq2$, a expressão

$$
\sqrt[n]{a}
$$

representa a raiz de índice $n$ de $a$.

Dizemos que

$$
\sqrt[n]{a}=b
$$

quando

$$
b^{n}=a.
$$

Na expressão $\sqrt[n]{a}$, o número $n$ é chamado de índice do radical, enquanto $a$ é chamado de radicando.<br><br>

Quando o índice é igual a $2$, costumamos omiti-lo:

$$
\sqrt[2]{a}=\sqrt{a}.
$$

</div>

Quando o índice do radical é par, o radicando deve ser maior ou igual a zero para que a raiz seja um número real. Por exemplo,

$$
\sqrt{16}=4,
$$

pois $4^{2}=16$. Entretanto, $\sqrt{-16}$ não representa um número real.

Quando o índice é ímpar, o radicando pode ser qualquer número real. Por exemplo,

$$
\sqrt[3]{-8}=-2,
$$

pois

$$
(-2)^{3}=-8.
$$

As principais propriedades da radiciação são apresentadas a seguir.

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

Para $a,b\in\mathbb{R}$ e $m,n\in\mathbb{N}$, com $m,n\geq2$, sempre que as expressões estiverem definidas, temos:<br><br>

$\sqrt[n]{ab}=\sqrt[n]{a}\cdot\sqrt[n]{b}$<br><br>
$\displaystyle\sqrt[n]{\frac{a}{b}}
=\frac{\sqrt[n]{a}}{\sqrt[n]{b}}$, com $b\neq0$<br><br>
$\sqrt[m]{\sqrt[n]{a}}=\sqrt[mn]{a}$<br><br>
$\displaystyle\sqrt[n]{a^{m}}=a^{\frac{m}{n}}$, sempre que a potência estiver definida no conjunto dos números reais<br><br>
$\displaystyle\sqrt[n]{a^{n}}
=
\begin{cases}
|a|, & \text{se }n\text{ for par},\\
a, & \text{se }n\text{ for ímpar}.
\end{cases}$

</div>

#### Exemplo 1

Use a definição de radiciação e as propriedades apresentadas acima para calcular ou simplificar as seguintes expressões:

a) $\sqrt{49}$

b) $\sqrt[3]{-27}$

c) $\sqrt{144}$

d) $\displaystyle\sqrt{\frac{25}{4}}$

e) $\sqrt{\sqrt[3]{64}}$

f) $\sqrt{(-5)^{2}}$

##### Solução do item a:

Como $49 = 7^{2}$, temos:

$$
\displaystyle \begin{aligned}
\sqrt{49}
&=\sqrt{7^{2}}\\
&=7^{\frac{2}{2}}\\
&=7^{1}\\
&=7
\end{aligned}
$$

##### Solução do item b:

Como

$$
(-3)^{3}=-27,
$$

temos, pela definição de raiz cúbica,

$$
\sqrt[3]{-27}=-3.
$$

##### Solução do item c:

Aplicando a propriedade da raiz de um produto e escrevendo os radicais na forma de potências, temos:

$$
\begin{aligned}
\sqrt{144}
&=\sqrt{2^{4}\cdot3^{2}}\\
&=\sqrt{2^{4}}\cdot\sqrt{3^{2}}\\
&=2^{\frac{4}{2}}\cdot3^{\frac{2}{2}}\\
&=2^{2}\cdot3^{1}\\
&=4\cdot3\\
&=12.
\end{aligned}
$$

##### Solução do item d:

Aplicando a propriedade da raiz de um quociente:

$$
\begin{aligned}
\sqrt{\frac{25}{4}}
&=\frac{\sqrt{25}}{\sqrt{4}}\\
&=\frac{5}{2}.
\end{aligned}
$$

##### Solução do item e:

Aplicando a propriedade da raiz de uma raiz:

$$
\begin{aligned}
\sqrt{\sqrt[3]{64}}
&=\sqrt[2\cdot3]{64}\\
&=\sqrt[6]{64}\\
&=\sqrt[6]{2^{6}}\\
&=2.
\end{aligned}
$$

##### Solução do item f:

Como o índice da raiz é par, temos:

$$
\begin{aligned}
\sqrt{(-5)^{2}}
&=|-5|\\
&=5.
\end{aligned}
$$

### 6.2.2 Extração de Fatores do Radical

Em algumas expressões, o radicando não é uma potência perfeita. Nesses casos, podemos fatorá-lo e retirar do radical os fatores que possuem raiz exata.

Por exemplo,

$$
\begin{aligned}
\sqrt{12}
&=\sqrt{4\cdot3}\\
&=\sqrt{4}\sqrt{3}\\
&=2\sqrt{3}.
\end{aligned}
$$

Esse procedimento é chamado de extração de fatores do radical.

#### Exemplo 1

Simplifique os seguintes radicais:

a) $\sqrt{72}$

b) $\sqrt{180}$

c) $\sqrt[3]{54}$

d) $\sqrt{48x^{2}}$, considerando $x\geq0$

##### Solução do item a:

Podemos começar escrevendo $72$ como um produto de fatores primos:

$$
72=2^{3}\cdot3^{2}.
$$

Como estamos trabalhando com uma raiz quadrada, separamos os fatores em grupos de expoente $2$:

$$
2^{3}=2^{2}\cdot2.
$$

Assim,

$$
\begin{aligned}
\sqrt{72}
&=\sqrt{2^{3}\cdot3^{2}}\\
&=\sqrt{2^{2}\cdot2\cdot3^{2}}\\
&=\sqrt{2^{2}}\cdot\sqrt2\cdot\sqrt{3^{2}}\\
&=2\cdot\sqrt2\cdot3\\
&=6\sqrt2.
\end{aligned}
$$

##### Solução do item b:

Como

$$
180=36\cdot5,
$$

temos:

$$
\begin{aligned}
\sqrt{180}
&=\sqrt{36\cdot5}\\
&=\sqrt{36}\cdot\sqrt5\\
&=6\sqrt5.
\end{aligned}
$$

##### Solução do item c:

Como

$$
54=3^{3}\cdot2,
$$

temos:

$$
\begin{aligned}
\sqrt[3]{54}
&=\sqrt[3]{3^{3}\cdot2}\\
&=\sqrt[3]{3^{3}}\cdot\sqrt[3]{2}\\
&=3\sqrt[3]{2}.
\end{aligned}
$$

##### Solução do item d:

Como $x\geq0$, temos $\sqrt{x^{2}}=x$. Assim,

$$
\begin{aligned}
\sqrt{48x^{2}}
&=\sqrt{16\cdot3\cdot x^{2}}\\
&=\sqrt{16}\cdot\sqrt3\cdot\sqrt{x^{2}}\\
&=4\sqrt3 \ x.
\end{aligned}
$$

Sem a condição $x\geq0$, o resultado deveria ser escrito como

$$
4\sqrt3 \ |x|.
$$

### 6.2.3 Operações com radicais

Radicais semelhantes são aqueles que possuem o mesmo índice e o mesmo radicando. Podemos adicionar ou subtrair os coeficientes de radicais semelhantes.

Por exemplo,

$$
2\sqrt3+5\sqrt3=7\sqrt3.
$$

Em alguns casos, devemos simplificar os radicais antes de realizar a adição ou a subtração.

#### Exemplo 3

Simplifique as seguintes expressões:

a) $3\sqrt5+2\sqrt5$

b) $\sqrt{12}+\sqrt{27}$

c) $2\sqrt8-\sqrt{18}$

d) $\sqrt8\cdot\sqrt6$

##### Solução do item a:

Como os radicais são semelhantes, adicionamos os coeficientes:

$$
\begin{aligned}
3\sqrt5+2\sqrt5
&=(3+2)\sqrt5\\
&=5\sqrt5.
\end{aligned}
$$

##### Solução do item b:

Inicialmente, simplificamos cada radical:

$$
\sqrt{12}=2\sqrt3
$$

e

$$
\sqrt{27}=3\sqrt3.
$$

Portanto,

$$
\begin{aligned}
\sqrt{12}+\sqrt{27}
&=2\sqrt3+3\sqrt3\\
&=5\sqrt3.
\end{aligned}
$$

##### Solução do item c:

Temos:

$$
\begin{aligned}
\sqrt8
&=\sqrt{4\cdot2}\\
&=\sqrt{4}\sqrt{2}\\
&=2\sqrt2.
\end{aligned}
$$

e

$$
\begin{aligned}
\sqrt18
&=\sqrt{9\cdot2}\\
&=\sqrt{9}\sqrt{2}\\
&=3\sqrt2.
\end{aligned}
$$

Assim,

$$
\begin{aligned}
2\sqrt8-\sqrt{18}
&=2\cdot2\sqrt2-3\sqrt2\\
&=4\sqrt2-3\sqrt2\\
&=\sqrt2.
\end{aligned}
$$

##### Solução do item d:

Aplicando a propriedade do produto de radicais:

$$
\begin{aligned}
\sqrt8\cdot\sqrt6
&=\sqrt{8\cdot6}\\
&=\sqrt{48}\\
&=\sqrt{16\cdot3}\\
&=4\sqrt3.
\end{aligned}
$$

### 6.2.4 Racionalização de Denominadores

Racionalizar o denominador de uma fração significa reescrevê-la de forma equivalente, eliminando os radicais presentes no denominador.

Esse procedimento será especialmente útil em algumas expressões que aparecem no estudo de limites.

#### Exemplo 4

Racionalize o denominador das seguintes expressões:

a) $\displaystyle\frac{1}{\sqrt2}$

b) $\displaystyle\frac{3}{2\sqrt5}$

##### Solução do item a:

Multiplicamos o numerador e o denominador por $\sqrt2$:

$$
\begin{aligned}
\frac{1}{\sqrt2}
&=\frac{1}{\sqrt2}\cdot\frac{\sqrt2}{\sqrt2}\\
&=\frac{\sqrt2}{(\sqrt2)^{2}}\\
&=\frac{\sqrt2}{2}.
\end{aligned}
$$

##### Solução do item b:

Multiplicamos o numerador e o denominador por $\sqrt5$:

$$
\begin{aligned}
\frac{3}{2\sqrt5}
&=\frac{3}{2\sqrt5}\cdot\frac{\sqrt5}{\sqrt5}\\
&=\frac{3\sqrt5}{2(\sqrt5)^{2}}\\
&=\frac{3\sqrt5}{10}.
\end{aligned}
$$

#### Exemplo 5

Racionalize o denominador das seguintes expressões:

a) $\displaystyle\frac{1}{2+\sqrt3}$

b) $\displaystyle\frac{1}{\sqrt{x+1}-1}$

##### Solução do item a:

O conjugado de $2+\sqrt3$ é $2-\sqrt3$. Multiplicamos o numerador e o denominador por essa expressão:

$$
\begin{aligned}
\frac{1}{2+\sqrt3}
&=
\frac{1}{2+\sqrt3}
\cdot
\frac{2-\sqrt3}{2-\sqrt3}\\
&=
\frac{2-\sqrt3}
{(2+\sqrt3)(2-\sqrt3)}\\
&=
\frac{2-\sqrt3}
{2^{2}-(\sqrt3)^{2}}\\
&=
\frac{2-\sqrt3}{4-3}\\
&=2-\sqrt3.
\end{aligned}
$$

##### Solução do item b:

O conjugado de $\sqrt{x+1}-1$ é $\sqrt{x+1}+1$. Assim,

$$
\begin{aligned}
\frac{1}{\sqrt{x+1}-1}
&=
\frac{1}{\sqrt{x+1}-1}
\cdot
\frac{\sqrt{x+1}+1}{\sqrt{x+1}+1}\\
&=
\frac{\sqrt{x+1}+1}
{\left(\sqrt{x+1}-1\right)\left(\sqrt{x+1}+1\right)}\\
&=
\frac{\sqrt{x+1}+1}
{(\sqrt{x+1})^{2}-1^{2}}\\
&=
\frac{\sqrt{x+1}+1}
{x+1-1}\\
&=
\frac{\sqrt{x+1}+1}{x}.
\end{aligned}
$$

Essa transformação é válida quando a expressão original está definida, isto é, quando

$$
x+1\geq0
$$

e

$$
\sqrt{x+1}-1\neq0.
$$

Portanto,

$$
x\geq-1
\qquad\text{e}\qquad
x\neq0.
$$

#### Exemplo 6

Simplifique a expressão:

$$
\frac{x-4}{\sqrt{x}-2}.
$$

##### Solução:

Multiplicamos o numerador e o denominador pelo conjugado de $\sqrt{x}-2$:

$$
\begin{aligned}
\frac{x-4}{\sqrt{x}-2}
&=
\frac{x-4}{\sqrt{x}-2}
\cdot
\frac{\sqrt{x}+2}{\sqrt{x}+2}\\
&=
\frac{(x-4)(\sqrt{x}+2)}
{(\sqrt{x}-2)(\sqrt{x}+2)}\\
&=
\frac{(x-4)(\sqrt{x}+2)}
{x-4}\\
&=
\sqrt{x}+2.
\end{aligned}
$$

O cancelamento é válido para $x\neq4$. Além disso, para que a raiz quadrada esteja definida, devemos ter $x\geq0$.

Assim, a expressão original está definida para

$$
x\geq0
\qquad\text{e}\qquad
x\neq4.
$$

Esse tipo de simplificação aparece frequentemente no cálculo de limites.

#### Exemplo 7:

Simplifique a expressão:

$$
\frac{x^{2}-25}{\sqrt{x+4}-3}.
$$

##### Solução:

Inicialmente, multiplicamos o numerador e o denominador pelo conjugado de $\sqrt{x+4}-3$, que é $\sqrt{x+4}+3$:

$$
\begin{aligned}
\frac{x^{2}-25}{\sqrt{x+4}-3}
&=
\frac{x^{2}-25}{\sqrt{x+4}-3}
\cdot
\frac{\sqrt{x+4}+3}{\sqrt{x+4}+3}\\
&=
\frac{(x^{2}-25)(\sqrt{x+4}+3)}
{(\sqrt{x+4}-3)(\sqrt{x+4}+3)}\\
&=
\frac{(x^{2}-25)(\sqrt{x+4}+3)}
{(\sqrt{x+4})^{2}-3^{2}}\\
&=
\frac{(x^{2}-25)(\sqrt{x+4}+3)}
{x+4-9}\\
&=
\frac{(x^{2}-25)(\sqrt{x+4}+3)}
{x-5}.
\end{aligned}
$$

Como o numerador é uma diferença de dois quadrados,

$$
x^{2}-25=(x-5)(x+5).
$$

Assim,

$$
\begin{aligned}
\frac{(x^{2}-25)(\sqrt{x+4}+3)}
{x-5}
&=
\frac{(x-5)(x+5)(\sqrt{x+4}+3)}
{x-5}\\
&=
(x+5)(\sqrt{x+4}+3).
\end{aligned}
$$

A expressão original está definida quando

$$
x+4\geq0
$$

e

$$
\sqrt{x+4}-3\neq0.
$$

Portanto,

$$
x\geq-4
\qquad\text{e}\qquad
x\neq5.
$$

Logo,

$$
\frac{x^{2}-25}{\sqrt{x+4}-3} = (x+5)(\sqrt{x+4}+3),
\qquad x\geq-4\text{ e }x\neq5.
$$

#### Exemplo 8

Racionalize o denominador da expressão:

$$
\frac{1}{1+\sqrt{2}+\sqrt{3}}.
$$

##### Solução:

Nesse caso, será necessário realizar a racionalização em duas etapas.

Inicialmente, agrupamos os dois primeiros termos do denominador:

$$
(1+\sqrt2)+\sqrt3.
$$

O conjugado dessa expressão, considerando os dois grupos, é

$$
1+\sqrt2-\sqrt3.
$$

Multiplicando o numerador e o denominador por esse conjugado, obtemos:

$$
\begin{aligned}
\frac{1}{1+\sqrt2+\sqrt3}
&=
\frac{1}{1+\sqrt2+\sqrt3}
\cdot
\frac{1+\sqrt2-\sqrt3}
{1+\sqrt2-\sqrt3}\\
&=
\frac{1+\sqrt2-\sqrt3}
{(1+\sqrt2+\sqrt3)(1+\sqrt2-\sqrt3)}\\
&=
\frac{1+\sqrt2-\sqrt3}
{(1+\sqrt2)^{2}-(\sqrt3)^{2}}\\
&=
\frac{1+\sqrt2-\sqrt3}
{1+2\sqrt2+2-3}\\
&=
\frac{1+\sqrt2-\sqrt3}{2\sqrt2}.
\end{aligned}
$$

O denominador ainda contém um radical. Por isso, realizamos uma segunda racionalização, multiplicando por $\sqrt2$:

$$
\begin{aligned}
\frac{1+\sqrt2-\sqrt3}{2\sqrt2}
&=
\frac{1+\sqrt2-\sqrt3}{2\sqrt2}
\cdot
\frac{\sqrt2}{\sqrt2}\\
&=
\frac{\sqrt2+2-\sqrt6}{4}\\
&=
\frac{2+\sqrt2-\sqrt6}{4}.
\end{aligned}
$$

## Exercícios

#### Exercício 1

Simplifique as expressões a seguir:

a) $\displaystyle \frac{3^{n+2}-3^{n}}{3^{n+1}}$

b) $\displaystyle \frac{2^{x+4}-2^{x+1}}{2^{x+2}+2^{x}}$

##### Solução do item a:

No numerador, colocamos $3^n$ em evidência:

$$
\begin{aligned}
3^{n+2}-3^n
&=3^n\cdot3^2-3^n\\
&=3^n(3^2-1)\\
&=3^n(9-1)\\
&=8\cdot3^n.
\end{aligned}
$$

Além disso,

$$
3^{n+1}=3^n\cdot3.
$$

Assim,

$$
\begin{aligned}
\frac{3^{n+2}-3^n}{3^{n+1}}
&=\frac{8\cdot3^n}{3^n\cdot3}\\
&=\frac{8}{3}.
\end{aligned}
$$

Como $3^n\neq0$, o cancelamento é válido.

##### Solução do item b:

No numerador, colocamos $2^{x+1}$ em evidência:

$$
\begin{aligned}
2^{x+4}-2^{x+1}
&=2^{x+1}\cdot2^3-2^{x+1}\\
&=2^{x+1}(2^3-1)\\
&=2^{x+1}(8-1)\\
&=7\cdot2^{x+1}.
\end{aligned}
$$

No denominador, colocamos $2^x$ em evidência:

$$
\begin{aligned}
2^{x+2}+2^x
&=2^x\cdot2^2+2^x\\
&=2^x(2^2+1)\\
&=2^x(4+1)\\
&=5\cdot2^x.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
\frac{2^{x+4}-2^{x+1}}
{2^{x+2}+2^x}
&=
\frac{7\cdot2^{x+1}}
{5\cdot2^x}\\
&=
\frac{7}{5}\cdot2^{x+1-x}\\
&=
\frac{7}{5}\cdot2\\
&=
\frac{14}{5}.
\end{aligned}
$$

Como $2^x\neq0$, o cancelamento é válido.

#### Exercício 3

Calcule o valor da expressão a seguir para $a=16$ e $b=8$:

$$
\frac{\left(a^{2}b^{-1}\right)^{3}}
{a^{3}\left(a^{-1}b\right)^{4}}.
$$

##### Solução:

Antes de substituir os valores de $a$ e $b$, podemos simplificar a expressão.

No numerador, temos:

$$
\begin{aligned}
\left(a^2b^{-1}\right)^3
&=a^{2\cdot3}b^{-1\cdot3}\\
&=a^6b^{-3}.
\end{aligned}
$$

No denominador,

$$
\begin{aligned}
a^3\left(a^{-1}b\right)^4
&=a^3\cdot a^{-4}b^4\\
&=a^{3-4}b^4\\
&=a^{-1}b^4.
\end{aligned}
$$

Assim,

$$
\begin{aligned}
\frac{a^6b^{-3}}{a^{-1}b^4}
&=a^{6-(-1)}b^{-3-4}\\
&=a^7b^{-7}\\
&=\frac{a^7}{b^7}\\
&=\left(\frac{a}{b}\right)^7.
\end{aligned}
$$

Substituindo $a=16$ e $b=8$, obtemos:

$$
\begin{aligned}
\left(\frac{a}{b}\right)^7
&=\left(\frac{16}{8}\right)^7\\
&=2^7\\
&=128.
\end{aligned}
$$

{7-1}\\

#### Exercício 9

Simplifique as expressões a seguir e determine suas condições de existência:

a) $\displaystyle\frac{x^{2}-16}{\sqrt{2x+1}-3}$

b) $\displaystyle\frac{x^{3}-8}{\sqrt{x+2}-2}$

##### Solução do item a:

Para que o radical esteja definido, devemos ter:

$$
2x+1\geq0,
$$

isto é,

$$
x\geq-\frac12.
$$

Além disso, o denominador deve ser diferente de zero:

$$
\sqrt{2x+1}-3\neq0.
$$

Logo,

$$
\sqrt{2x+1}\neq3
$$

e, portanto,

$$
x\neq4.
$$

Multiplicamos o numerador e o denominador pelo conjugado $\sqrt{2x+1}+3$:

$$
\begin{aligned}
\frac{x^2-16}{\sqrt{2x+1}-3}
&=
\frac{(x^2-16)(\sqrt{2x+1}+3)}
{(\sqrt{2x+1}-3)(\sqrt{2x+1}+3)}\\
&=
\frac{(x^2-16)(\sqrt{2x+1}+3)}
{2x+1-9}\\
&=
\frac{(x^2-16)(\sqrt{2x+1}+3)}
{2x-8}.
\end{aligned}
$$

Fatorando,

$$
x^2-16=(x-4)(x+4)
$$

e

$$
2x-8=2(x-4).
$$

Assim,

$$
\begin{aligned}
\frac{(x^2-16)(\sqrt{2x+1}+3)}
{2x-8}
&=
\frac{(x-4)(x+4)(\sqrt{2x+1}+3)}
{2(x-4)}\\
&=
\frac{(x+4)(\sqrt{2x+1}+3)}{2}.
\end{aligned}
$$

A expressão simplificada é

$$
\frac{(x+4)(\sqrt{2x+1}+3)}{2},
\qquad
x\geq-\frac12
\quad\text{e}\quad
x\neq4.
$$

##### Solução do item b:

Para que o radical esteja definido, devemos ter:

$$
x+2\geq0,
$$

isto é,

$$
x\geq-2.
$$

Além disso,

$$
\sqrt{x+2}-2\neq0,
$$

de modo que

$$
x\neq2.
$$

Multiplicamos o numerador e o denominador pelo conjugado $\sqrt{x+2}+2$:

$$
\begin{aligned}
\frac{x^3-8}{\sqrt{x+2}-2}
&=
\frac{(x^3-8)(\sqrt{x+2}+2)}
{(\sqrt{x+2}-2)(\sqrt{x+2}+2)}\\
&=
\frac{(x^3-8)(\sqrt{x+2}+2)}
{x+2-4}\\
&=
\frac{(x^3-8)(\sqrt{x+2}+2)}
{x-2}.
\end{aligned}
$$

Como o numerador é uma diferença de dois cubos,

$$
x^3-8=(x-2)(x^2+2x+4).
$$

Portanto,

$$
\begin{aligned}
\frac{(x^3-8)(\sqrt{x+2}+2)}
{x-2}
&=
\frac{(x-2)(x^2+2x+4)(\sqrt{x+2}+2)}
{x-2}\\
&=
(x^2+2x+4)(\sqrt{x+2}+2).
\end{aligned}
$$

A expressão simplificada é

$$
(x^2+2x+4)(\sqrt{x+2}+2),
\qquad
x\geq-2
\quad\text{e}\quad
x\neq2.
$$

#### Exercício 10

Racionalize completamente o denominador da expressão:

$$
\frac{1}{\sqrt2+\sqrt3+\sqrt5}.
$$

##### Solução:

Inicialmente, agrupamos os dois primeiros termos do denominador:

$$
(\sqrt2+\sqrt3)+\sqrt5.
$$

Multiplicamos o numerador e o denominador por

$$
\sqrt2+\sqrt3-\sqrt5.
$$

Assim,

$$
\begin{aligned}
\frac{1}{\sqrt2+\sqrt3+\sqrt5}
&=
\frac{\sqrt2+\sqrt3-\sqrt5}
{(\sqrt2+\sqrt3+\sqrt5)
(\sqrt2+\sqrt3-\sqrt5)}\\
&=
\frac{\sqrt2+\sqrt3-\sqrt5}
{(\sqrt2+\sqrt3)^2-(\sqrt5)^2}\\
&=
\frac{\sqrt2+\sqrt3-\sqrt5}
{2+3+2\sqrt6-5}\\
&=
\frac{\sqrt2+\sqrt3-\sqrt5}
{2\sqrt6}.
\end{aligned}
$$

Como o denominador ainda contém um radical, multiplicamos o numerador e o denominador por $\sqrt6$:

$$
\begin{aligned}
\frac{\sqrt2+\sqrt3-\sqrt5}{2\sqrt6}
&=
\frac{\sqrt6(\sqrt2+\sqrt3-\sqrt5)}
{2\sqrt6\cdot\sqrt6}\\
&=
\frac{\sqrt{12}+\sqrt{18}-\sqrt{30}}
{12}\\
&=
\frac{2\sqrt3+3\sqrt2-\sqrt{30}}
{12}.
\end{aligned}
$$
