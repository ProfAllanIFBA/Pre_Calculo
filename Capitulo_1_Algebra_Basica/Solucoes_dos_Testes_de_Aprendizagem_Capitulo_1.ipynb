# Teste de Aprendizagem da Aula 1

#### Questão 1

Assinale a igualdade que não é verdadeira:

a) $6+(4+9)=(6+4)+9$

b) $3\cdot(5-2)=3\cdot5-3\cdot2$

c) $24\div(6\div2)=(24\div6)\div2$

d) $7\cdot4=4\cdot7$

e) Todas as igualdades anteriores são verdadeiras.

##### Solução:

Analisando cada igualdade, temos:

Alternativa a)

$$
\begin{aligned}
6+(4+9)
&=6+13\
&=19,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
(6+4)+9
&=10+9\
&=19.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade associativa da adição.

Alternativa b)

$$
\begin{aligned}
3\cdot(5-2)
&=3\cdot3\
&=9,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
3\cdot5-3\cdot2
&=15-6\
&=9.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade distributiva.

Alternativa c)

$$
\begin{aligned}
24\div(6\div2)
&=24\div3\
&=8,
\end{aligned}
$$

enquanto

$$
\begin{aligned}
(24\div6)\div2
&=4\div2\
&=2.
\end{aligned}
$$

Portanto,

$$
24\div(6\div2)\neq(24\div6)\div2.
$$

Logo, a igualdade não é verdadeira, pois a divisão não é uma operação associativa.

Alternativa d)

$$
\begin{aligned}
7\cdot4&=28,\
4\cdot7&=28.
\end{aligned}
$$

Portanto, a igualdade é verdadeira. Ela representa a propriedade comutativa da multiplicação.

Assim, a alternativa correta é c).

#### Questão 1

O valor da expressão

$$
\frac{1}{2}-\frac{1}{2^{2}}+\frac{1}{2^{3}}
$$

é:

a) $1$

b) $\frac{3}{8}$

c) $\frac{3}{4}$

d) $\frac{4}{3}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Calculamos inicialmente as potências:

$$
2^2=4
\ \ \text{e} \ \
2^3=8.
$$

Assim,

$$
\begin{aligned}
\frac{1}{2}-\frac{1}{2^2}+\frac{1}{2^3}
&=\frac{1}{2}-\frac{1}{4}+\frac{1}{8}\\
&=\frac{4}{8}-\frac{2}{8}+\frac{1}{8}\\
&=\frac{4-2+1}{8}\\
&=\frac{3}{8}.
\end{aligned}
$$

Portanto, a alternativa correta é b).

\left(\frac{5}{8}\cdot\frac{12}{25}-\frac{1}{10}\right)^{2}

c) $\frac{2}{15}$

d) $\frac{1}{25}$

e) Nenhuma das alternativas anteriores está correta.

$$
\begin{aligned}
\frac{5}{8}\cdot\frac{12}{25}-\frac{1}{10}

&=\frac{3}{10}-\frac{1}{10}\\

Agora, elevamos o resultado ao quadrado:

$$\frac{1^2}{5^2}= \frac{1}{25}.$$

Portanto, a alternativa correta é d).

#### Questão 4

Considere as variáveis:

$$ a= 1 + \frac{1}{1+1}, \ \ b = 1 + \frac{1}{1+\frac{1}{1+1}}, \ \ c = 1 + \frac{1}{1+\frac{1}{1+\frac{1}{1+1}}}$$

O valor do produto $abc$ é dado por:

a) $4$

b) $3$

c) $\frac{3}{4}$

d) $\frac{2}{3}$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Calculamos inicialmente o valor de $a$:

$$
\begin{aligned}
a
&=1+\frac{1}{1+1}\\
&=1+\frac{1}{2}\\
&=\frac{3}{2}.
\end{aligned}
$$

Calculamos agora o valor de $b$, começando pelas operações mais internas:

$$
\begin{aligned}
b
&=1+\frac{1}{1+\frac{1}{1+1}}\\
&=1+\frac{1}{1+\frac{1}{2}}\\
&=1+\frac{1}{\frac{3}{2}}\\
&=1+\frac{2}{3}\\
&=\frac{5}{3}.
\end{aligned}
$$

De modo semelhante, calculamos o valor de $c$

$$
\begin{aligned}
c
&=1+\frac{1}{1+\frac{1}{1+\frac{1}{1+1}}}\\
&=1+\frac{1}{1+\frac{1}{1+\frac{1}{2}}}\\
&=1+\frac{1}{1+\frac{1}{\frac{3}{2}}}\\
&=1+\frac{1}{1+\frac{2}{3}}\\
&=1+\frac{1}{\frac{5}{3}}\\
&=1+\frac{3}{5}\\
&=\frac{8}{5}.
\end{aligned}
$$

Portanto,

$$
\begin{aligned}
abc
&=\frac{3}{2}\cdot\frac{5}{3}\cdot\frac{8}{5}\\
&=\frac{3\cdot5\cdot8}{2\cdot3\cdot5}\\
&=\frac{8}{2}\\
&=4.
\end{aligned}
$$

Logo, a alternativa correta é a).

# Teste de Aprendizagem da Aula 3

#### Questão 1

Considere as expressões algébricas

$$
3(x+y)
\ \ \text{e} \ \
3x+xy.
$$

Qual dos pares de valores a seguir constitui um contraexemplo para a afirmação de que essas expressões são equivalentes?

a) $x=0$ e $y=0$

b) $x=1$ e $y=0$

c) $x=-2$ e $y=0$

d) $x=3$ e $y=1$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Vamos verificar os pares de valores apresentados.

Para $x=0$ e $y=0$:

$$
3(x+y)=3(0+0)=0
$$

e

$$
3x+xy=3\cdot0+0\cdot0=0.
$$

Para $x=1$ e $y=0$:

$$
3(x+y)=3(1+0)=3
$$

e

$$
3x+xy=3\cdot1+1\cdot0=3.
$$

Para $x=-2$ e $y=0$:

$$
3(x+y)=3(-2+0)=-6
$$

e

$$
3x+xy=3\cdot(-2)+(-2)\cdot0=-6.
$$

Para $x=3$ e $y=1$:

$$
3(x+y)=3(3+1)=12
$$

e

$$
3x+xy=3\cdot3+3\cdot1=12.
$$

Portanto, em todos os pares apresentados, as expressões assumem o mesmo valor. Logo, nenhuma das alternativas de a) a d) constitui um contraexemplo.

A alternativa correta é:

e) Nenhuma das alternativas anteriores está correta.

Podemos compreender de forma geral por que isso ocorre. Calculando a diferença entre as expressões, temos:

$$
\begin{aligned}
3(x+y)-(3x+xy)
&=3x+3y-3x-xy\\
&=3y-xy\\
&=y(3-x).
\end{aligned}
$$

As expressões assumem o mesmo valor quando

$$
y(3-x)=0,
$$

isto é, quando

$$
y=0
\ \ \text{ou} \ \
x=3.
$$

Nas alternativas a), b) e c), temos $y=0$. Na alternativa d), temos $x=3$. Por isso, todas produzem resultados iguais.

Entretanto, as expressões não são equivalentes. Considerando, por exemplo, $x=1$ e $y=1$, obtemos:

$$
3(x+y)=3(1+1)=6
$$

e

$$
3x+xy=3\cdot1+1\cdot1=4.
$$

Assim,

$$
3(x+y)\neq3x+xy
$$

quando $x=1$ e $y=1$. Portanto, esse par é um contraexemplo correto e mostra que as expressões não são equivalentes.

#### Questão 4

Considere a expressão

$$
\frac{ax+ay+bx+by}{x^2+xy}.
$$

Sua forma simplificada, acompanhada das restrições da expressão original, é:

a) $\displaystyle\frac{a+b}{x}$, com $x\neq0$ e $x\neq-y$.

b) $\displaystyle\frac{a+b}{x+y}$, com $x\neq0$.

c) $\displaystyle\frac{a+b}{x}$, com apenas $x\neq0$.

d) $\displaystyle\frac{a+b}{x+y}$, com $x\neq-y$.

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Fatorando o numerador e o denominador, temos:

$$
\begin{aligned}
\frac{ax+ay+bx+by}{x^2+xy}
&=\frac{a(x+y)+b(x+y)}{x(x+y)}\\
&=\frac{(a+b)(x+y)}{x(x+y)}\\
&=\frac{a+b}{x}.
\end{aligned}
$$

O denominador da expressão original pode ser escrito como

$$
x^2+xy=x(x+y).
$$

Para que a expressão original esteja definida, devemos ter:

$$
x\neq0
\ \ \text{e} \ \
x+y\neq0.
$$

Equivalentemente,

$$
x\neq0
\ \ \text{e} \ \
x\neq-y.
$$

Portanto,

$$
\frac{ax+ay+bx+by}{x^2+xy} = \frac{a+b}{x},
$$

desde que

$$
x\neq0
\qquad\text{e}\qquad
x\neq-y.
$$

A alternativa correta é a).

Considere as expressões

$$
\text{A}=(4x+3)^{2}
\qquad\text{e}\qquad
\text{B}=2(a-2b)(a+2b).
$$

As formas desenvolvidas de $\text{A}$ e $\text{B}$ são, respectivamente:

a) $\text{A}=16x^{2}+9$ e $\text{B}=2a^{2}-8b^{2}$;

b) $\text{A}=16x^{2}+24x+9$ e $\text{B}=2a^{2}-8b^{2}$;

c) $\text{A}=16x^{2}+12x+9$ e $\text{B}=2a^{2}-4b^{2}$;

d) $\text{A}=8x^{2}+24x+9$ e $\text{B}=2a^{2}+8b^{2}$;

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Para desenvolver a expressão $\text{A}$, utilizamos a fórmula

$$
(a+b)^{2}=a^{2}+2ab+b^{2}.
$$

Assim,

$$
\begin{aligned}
\text{A}
&=(4x+3)^{2}\\
&=(4x)^{2}+2\cdot(4x)\cdot3+3^{2}\\
&=16x^{2}+24x+9.
\end{aligned}
$$

Na expressão $\text{B}$, utilizamos o produto da soma pela diferença:

$$
(a-b)(a+b)=a^{2}-b^{2}.
$$

Logo,

$$
\begin{aligned}
\text{B}
&=2(a-2b)(a+2b)\\
&=2\left[a^{2}-(2b)^{2}\right]\\
&=2(a^{2}-4b^{2})\\
&=2a^{2}-8b^{2}.
\end{aligned}
$$

Portanto, a alternativa correta é b).

#### Questão 4

A expressão

$$
\frac{x^{2}+x-6}{x^{2}-4}
$$

pode ser simplificada para:

a) $\displaystyle \frac{x+3}{x-2}$, com $x\neq-2$ e $x\neq2$;

b) $\displaystyle \frac{x+3}{x+2}$, apenas com $x\neq-2$;

c) $\displaystyle \frac{x-3}{x+2}$, com $x\neq-2$ e $x\neq2$;

d) $\displaystyle \frac{x+3}{x+2}$, com $x\neq-2$ e $x\neq2$;

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Inicialmente, fatoramos o numerador. Para a equação

$$
x^{2}+x-6=0,
$$

temos

$$
a=1,\qquad b=1,\qquad c=-6.
$$

Calculando o discriminante:

$$
\begin{aligned}
\Delta
&=b^{2}-4ac\\
&=1^{2}-4\cdot1\cdot(-6)\\
&=1+24\\
&=25.
\end{aligned}
$$

As raízes são

$$
\begin{aligned}
x
&=\frac{-b\pm\sqrt{\Delta}}{2a}\\
&=\frac{-1\pm\sqrt{25}}{2\cdot1}\\
&=\frac{-1\pm5}{2}.
\end{aligned}
$$

Portanto,

$$
x_{1}=2
\qquad\text{e}\qquad
x_{2}=-3.
$$

Logo,

$$
x^{2}+x-6=(x-2)(x+3).
$$

Agora, fatoramos o denominador utilizando a diferença de dois quadrados:

$$
\begin{aligned}
x^{2}-4
&=x^{2}-2^{2}\\
&=(x-2)(x+2).
\end{aligned}
$$

Substituindo as fatorações na expressão, obtemos

$$
\begin{aligned}
\frac{x^{2}+x-6}{x^{2}-4}
&=\frac{(x-2)(x+3)}{(x-2)(x+2)}\\
&=\frac{x+3}{x+2}.
\end{aligned}
$$

Entretanto, o denominador da expressão original não pode ser igual a zero. Assim,

$$
\begin{aligned}
x^{2}-4&\neq0\\
(x-2)(x+2)&\neq0.
\end{aligned}
$$

Portanto,

$$
x\neq2
\qquad\text{e}\qquad
x\neq-2.
$$

Logo,

$$
\frac{x^{2}+x-6}{x^{2}-4} = \frac{x+3}{x+2},
\qquad x\neq-2\text{ e }x\neq2.
$$

A alternativa correta é d).

#### Questão 5

Para $h\neq0$, a expressão

$$
\frac{(3x+h)^{2}-9x^{2}}{h}
$$

é equivalente a:

a) $3x+h$;

b) $9x+h$;

c) $6x+h^{2}$;

d) $6x+h$;

e) Nenhuma das alternativas anteriores está correta.

#### Solução:

Inicialmente, desenvolvemos o quadrado do binômio:

$$
\begin{aligned}
(3x+h)^{2}
&=(3x)^{2}+2\cdot(3x)\cdot h+h^{2}\\
&=9x^{2}+6xh+h^{2}.
\end{aligned}
$$

Substituindo esse resultado na expressão:

$$
\begin{aligned}
\frac{(3x+h)^{2}-9x^{2}}{h}
&=\frac{9x^{2}+6xh+h^{2}-9x^{2}}{h}\\
&=\frac{6xh+h^{2}}{h}\\
&=\frac{h(6x+h)}{h}\\
&=6x+h.
\end{aligned}
$$

Como o denominador da expressão original é $h$, a condição $h\neq0$ deve ser mantida.

Portanto, a alternativa correta é d).

A fatoração correta da expressão

$$
27x^{3}+8
$$

é:

a) $(27x+8)(x^2-x+1)$

b) $(3x+2)(9x^2+6x+4)$

c) $(3x-2)(9x^2+6x+4)$

d) $(3x+2)(9x^2-4)$

##### Solução:

A expressão é uma soma de dois cubos, pois

$$
27x^3=(3x)^3
\qquad\text{e}\qquad
8=2^3.
$$

Utilizando a fórmula

$$
a^3+b^3=(a+b)(a^2-ab+b^2),
$$

obtemos

$$
\begin{aligned}
27x^3+8
&=(3x)^3+2^3\\
&=(3x+2)\left((3x)^2-(3x)\cdot2+2^2\right)\\
&=(3x+2)(9x^2-6x+4).
\end{aligned}
$$

#### Questão 5

Simplifique a expressão e determine suas condições de existência:

$$
\frac{x^{2}-9}{\sqrt{x+6}-3}.
$$

A alternativa correta é:

a) $(x+3)(\sqrt{x+6}+3)$, com $x\geq-6$ e $x\neq3$

b) $(x-3)(\sqrt{x+6}+3)$, com $x\geq-6$ e $x\neq-3$

d) $x+3$, com $x\geq-6$ e $x\neq3$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Primeiramente, precisamos considerar as condições de existência da expressão.

Como temos uma raiz quadrada, devemos ter:

$$
x+6\geq0.
$$

Logo,

$$
x\geq-6.
$$

Além disso, o denominador não pode ser igual a zero:

$$
\sqrt{x+6}-3\neq0.
$$

Assim,

$$
\sqrt{x+6}\neq3.
$$

Portanto,

$$
x+6\neq9.
$$

Logo,

$$
x\neq3.
$$

Agora, racionalizamos o denominador multiplicando o numerador e o denominador por $\sqrt{x+6}+3$:

$$
\frac{x^{2}-9}{\sqrt{x+6}-3}
=
\frac{x^{2}-9}{\sqrt{x+6}-3}
\cdot
\frac{\sqrt{x+6}+3}{\sqrt{x+6}+3}.
$$

\frac{x^{2}-9}{\sqrt{x+6}-3}
=
\frac{(x^{2}-9)(\sqrt{x+6}+3)}{(\sqrt{x+6}-3)(\sqrt{x+6}+3)}.

(\sqrt{x+6}-3)(\sqrt{x+6}+3)

(\sqrt{x+6})^2-3^2
=
x+6-9

\frac{x^{2}-9}{\sqrt{x+6}-3}
