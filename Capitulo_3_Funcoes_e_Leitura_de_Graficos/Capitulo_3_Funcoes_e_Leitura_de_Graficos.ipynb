## 1.1 Relações entre Grandezas e Conceito de Função


Em muitas situações, o valor de uma grandeza depende do valor de outra.

Por exemplo:

- o valor pago em uma corrida por aplicativo depende da distância percorrida;
- a quantidade de água em um reservatório pode depender do tempo;
- a área de um quadrado depende da medida de seu lado;
- a temperatura de uma substância pode depender do tempo de aquecimento.

Uma **função** é uma maneira de representar matematicamente uma relação desse tipo.


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
Uma função $f:A\to B$ associa a <strong>cada</strong> elemento $x$ do conjunto $A$ <strong>um único</strong> elemento $y$ do conjunto $B$.

Usualmente escrevemos

$$
y=f(x).
$$

A variável $x$ é chamada de <strong>variável independente</strong> ou variável de entrada, enquanto $y=f(x)$ é a <strong>variável dependente</strong> ou variável de saída.
</div>


É importante observar duas ideias:

1. todo valor permitido de entrada deve possuir uma saída;
2. uma mesma entrada não pode produzir duas saídas diferentes.

Entretanto, **entradas diferentes podem produzir a mesma saída**.


#### Exemplo 1


Um estacionamento cobra uma taxa fixa de $R\$ 6,00$, acrescida de $R\$ 3,00$ por hora de permanência. Se $t$ representa o número de horas de permanência, o custo pode ser representado por

$$
C(t)=6+3t.
$$

Determine $C(0)$, $C(1)$ e $C(3)$.

##### Soluções:


A notação $C(t)$ indica que o custo $C$ depende do tempo de permanência $t$.

Para determinar o custo correspondente a um determinado tempo, substituímos $t$ pelo valor desejado.

Para $t=0$:

$$
C(0)=6+3\cdot0=6.
$$

Para $t=1$:

$$
C(1)=6+3\cdot1=9.
$$

Para $t=3$:

$$
C(3)=6+3\cdot3=15.
$$

Portanto,

$$
C(0)=6,\qquad C(1)=9
\qquad\text{e}\qquad
C(3)=15.
$$

Por exemplo, $C(3)=15$ significa que, para uma permanência de $3$ horas, o custo é de R\$ 15,00.

De maneira geral, quando uma função é escrita na forma

$$
y=f(x),
$$

o valor $f(a)$ é obtido substituindo $x$ por $a$ na expressão que define a função.

#### Exemplo 2


Considere a relação dada pela tabela:

$$
\begin{array}{c|ccccc}
x & -2 & -1 & 0 & 1 & 2 \\ \hline
y & 4 & 1 & 0 & 1 & 4
\end{array}
$$

Essa relação representa uma função de $x$?

##### Solução:


Sim. Cada valor de $x$ aparece associado a **um único** valor de $y$.

Observe que $-2$ e $2$ possuem a mesma saída, pois ambos estão associados a $4$. Isso não é um problema: uma função pode atribuir a mesma saída a entradas diferentes.

O que não pode ocorrer é uma mesma entrada possuir duas saídas distintas.


#### Exemplo 3


Considere agora a relação

$$
\{(0,1),(1,2),(1,4),(2,5)\}.
$$

Ela representa uma função de $x$?


##### Solução:


Não. A entrada $x=1$ aparece associada a dois valores diferentes:

$$
1\longmapsto2
\qquad\text{e}\qquad
1\longmapsto4.
$$

Portanto, essa relação não pode representar uma função de $x$.


#### Exemplo 4


Considere

$$
g(x)=2x-5.
$$

Determine o valor de $x$ para o qual $g(x)=9$.

##### Solução:


Neste caso, não estamos calculando diretamente uma saída. Estamos procurando a entrada que produz a saída $9$.

Assim,

$$
2x-5=9.
$$

Logo,

$$
2x=14
\quad\Longrightarrow\quad
x=7.
$$

Portanto,

$$
g(7)=9.
$$


#### Exemplo 5

Considere

$$
h(x)=x^{2}+4x.
$$

Determine $x$ de modo que $h(x)=12$.

##### Solução:

Para determinar $x$, fazemos

$$
x^2+4x=12.
$$

Passando todos os termos para o primeiro membro:

$$
x^2+4x-12=0.
$$

Fatorando:

$$
(x+6)(x-2)=0.
$$

Pela propriedade do produto nulo,

$$
x+6=0
\qquad\text{ou}\qquad
x-2=0.
$$

Logo,

$$
x=-6
\qquad\text{ou}\qquad
x=2.
$$

Portanto, nesse caso, **duas entradas diferentes produzem a mesma saída $12$**:

$$
h(-6)=12
\qquad\text{e}\qquad
h(2)=12.
$$

## 1.2 Domínio, Contradomínio e Imagem

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

Considere uma função escrita como

$$
\color{red}{f:A\to B.}
$$

O conjunto $A$ é o conjunto de entrada, enquanto o conjunto $B$ é o conjunto de chegada da função.

<ul>
<li>O <strong>domínio</strong> é o conjunto de todas as entradas da função.</li>
<li>O <strong>contradomínio</strong> é o conjunto de chegada da função.</li>
<li>A <strong>imagem</strong> é o conjunto dos valores do contradomínio que são efetivamente produzidos pela função.</li>
</ul>

Em particular,

$$
\color{red}{\operatorname{Im}(f)\subseteq B.}
$$

Isso significa que nem todo elemento do contradomínio precisa pertencer à imagem da função.

</div>

#### Exemplo 6


Considere

$$
f:\{-2,-1,0,1,2\}\to\{0,1,4,9\},
\qquad
f(x)=x^2.
$$

Determine o domínio, o contradomínio e a imagem.


##### Solução:


O domínio é o conjunto das entradas:

$$
D(f)=\{-2,-1,0,1,2\}.
$$

O contradomínio é o conjunto indicado após a seta:

$$
\{0,1,4,9\}.
$$

Calculando as saídas:

$$
f(-2)=4,\quad
f(-1)=1,\quad
f(0)=0,\quad
f(1)=1,\quad
f(2)=4.
$$

Logo,

$$
\operatorname{Im}(f)=\{0,1,4\}.
$$

O valor $9$ pertence ao contradomínio, mas não pertence à imagem, pois nenhuma entrada do domínio produz esse valor.


<div style="background-color:#fff7e6; color:#333333; border-left:5px solid #f0ad4e; padding:10px; border-radius:6px;">
<strong>Cuidado:</strong> Quando uma função é apresentada apenas por uma fórmula e o domínio não é informado, consideramos, neste curso, <b>o maior subconjunto dos números reais para o qual a expressão está definida</b>.

As situações mais frequentes que devemos observar são:

- denominadores devem ser diferentes de zero;
- em raízes de índice par, o radicando deve ser maior ou igual a zero;
- em raízes de índice ímpar, o radicando pode assumir qualquer valor real.
</div>

#### Exemplo 7


Determine o domínio de cada função:

a) $f(x)=x^2-4x+1$

b) $\displaystyle g(x)=\frac{x+2}{x-3}$

c) $\displaystyle h(x)=\sqrt{2x-4}$


##### Solução do item a:


A expressão é polinomial e está definida para todo número real. Portanto,

$$
D(f)=\mathbb{R}.
$$


##### Solução do item b:


O denominador não pode ser igual a zero:

$$
x-3\neq0.
$$

Assim,

$$
x\neq3
$$

e

$$
D(g)=\mathbb{R}\setminus\{3\}.
$$


##### Solução do item c:


Como a raiz é quadrada, devemos ter

$$
2x-4\geq0.
$$

Logo,

$$
2x\geq4
\quad\Longrightarrow\quad
x\geq2.
$$

Portanto,

$$
D(h)=[2,+\infty).
$$


#### Exemplo 8


Determine o domínio da função

$$
q(x)=\frac{\sqrt{x+1}}{x-2}.
$$


##### Solução:


Devemos satisfazer simultaneamente duas condições.

Para a raiz:

$$
x+1\geq0
\quad\Longrightarrow\quad
x\geq-1.
$$

Para o denominador:

$$
x-2\neq0
\quad\Longrightarrow\quad
x\neq2.
$$

Combinando as condições:

$$
D(q)=[-1,2)\cup(2,+\infty).
$$


##### Observação


O domínio deve ser determinado a partir da **expressão original**.

Por exemplo,

$$
r(x)=\frac{x^2-1}{x-1}
$$

pode ser simplificada, para $x\neq1$, como

$$
r(x)=x+1.
$$

Observe que a expressão simplificada $x+1$ está definida em $x=1$. Entretanto, a expressão original

$$
\frac{x^2-1}{x-1}
$$

não está definida nesse valor, pois seu denominador é igual a zero.

Portanto,

$$
D(r)=\mathbb{R}\setminus\{1\}.
$$

A simplificação não faz o valor excluído voltar ao domínio.

#### Exemplo 9

Determine o domínio da função

$$
f(x)=\frac{1}{x}+\frac{1}{x-1}.
$$

##### Solução:

Para que a função esteja definida, seus denominadores devem ser diferentes de zero.

Assim,

$$
x\neq0
$$

e

$$
x-1\neq0
\quad\Longrightarrow\quad
x\neq1.
$$

Portanto, devemos excluir os valores $0$ e $1$ do conjunto dos números reais.

Logo,

$$
D(f)=\mathbb{R}\setminus\{0,1\}.
$$

## 1.3 Formas de Representar uma Função


Uma mesma função pode ser apresentada de diferentes maneiras. Entre as mais comuns estão:

- uma **fórmula**;
- uma **tabela**;
- um conjunto de **pares ordenados**;
- um **gráfico**;
- uma descrição em linguagem comum.

Essas representações fornecem informações diferentes sobre o mesmo objeto matemático.


#### Exemplo 10


Considere

$$
f(x)=2x+1
$$

com domínio

$$
\{-2,-1,0,1,2\}.
$$

A função pode ser representada pela tabela:

$$
\begin{array}{c|ccccc}
x & -2 & -1 & 0 & 1 & 2 \\ \hline
y & -3 & -1 & 1 & 3 & 5
\end{array}
$$

ou pelo conjunto de pares ordenados

$$
\{(-2,-3),(-1,-1),(0,1),(1,3),(2,5)\}.
$$

Os mesmos pares podem ser marcados no plano cartesiano.<br><br>

## 1.4 Quando um Gráfico Representa uma Função?


Em um gráfico de uma função de $x$, uma mesma coordenada horizontal não pode corresponder a duas alturas diferentes.

Isso fornece um teste visual simples.


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">
<strong>Teste da reta vertical:</strong> se alguma reta vertical interceptar o gráfico em mais de um ponto, então o gráfico não representa uma função de $x$.
</div>


#### Exemplo 11


A relação

$$
x^2+y^2=4
$$

representa uma circunferência. Ela define $y$ como função de $x$?<br><br>

<center> <img src="img/fig2.jpg" width="400"> </center>

##### Solução:


Não.

Por exemplo, para $x=0$, a relação fornece

$$
0^2+y^2=4,
$$

isto é,

$$
y=2
\qquad\text{ou}\qquad
y=-2.
$$

Assim, o mesmo valor $x=0$ está associado a dois valores diferentes de $y$. Geometricamente, a reta vertical $x=0$ corta a circunferência em dois pontos.


## Exercícios


#### Exercício 1


Considere

$$
f(x)=x^2-2x-3.
$$

Determine:

a) $f(-1)$

b) $f(2)+f(-2)\cdot f(3)$

c) o valor de $x$ para o qual $f(x)=-4$.

##### Solução do item a:

Substituindo $x=-1$:

$$
\begin{aligned}
f(-1)
&=(-1)^2-2(-1)-3\\
&=1+2-3\\
&=0.
\end{aligned}
$$

##### Solução do item b:

Calculamos inicialmente os valores da função:

$$
f(2)=2^2-2\cdot2-3=-3,
$$

$$
f(-2)=(-2)^2-2(-2)-3=5,
$$

e

$$
f(3)=3^2-2\cdot3-3=0.
$$

Portanto,

$$
\begin{aligned}
f(2)+f(-2)\cdot f(3)
&=-3+5\cdot0\\
&=-3.
\end{aligned}
$$

##### Solução do item c:

Queremos determinar $x$ de modo que

$$
f(x)=-4.
$$

Assim,

$$
x^2-2x-3=-4.
$$

Logo,

$$
x^2-2x+1=0.
$$

Fatorando,

$$
(x-1)^2=0.
$$

Portanto,

$$
x=1.
$$

#### Exercício 2

Considere a função

$$
f(x)=x^2-4
$$

com domínio

$$
\{-2,-1,0,1,2\}.
$$

Represente essa função por meio de:

a) uma tabela;

b) um conjunto de pares ordenados;

c) um gráfico no plano cartesiano.

##### Solução do item a:

Calculamos o valor da função para cada elemento do domínio.

Para $x=-2$:

$$
f(-2)=(-2)^2-4=0.
$$

Para $x=-1$:

$$
f(-1)=(-1)^2-4=-3.
$$

Para $x=0$:

$$
f(0)=0^2-4=-4.
$$

Para $x=1$:

$$
f(1)=1^2-4=-3.
$$

Para $x=2$:

$$
f(2)=2^2-4=0.
$$

Assim, obtemos a tabela

$$
\begin{array}{c|ccccc}
x & -2 & -1 & 0 & 1 & 2 \\ \hline
f(x) & 0 & -3 & -4 & -3 & 0
\end{array}
$$

##### Solução do item b:

Com base nos valores obtidos no item a, o conjunto de pares ordenados é

$$
\{(-2,0),(-1,-3),(0,-4),(1,-3),(2,0)\}.
$$

##### Solução do item c:

Com base nos pares ordenados obtidos no item b, marcamos os pontos correspondentes no plano cartesiano e obtemos o gráfico da função no domínio dado.<br><br>

#### Exercício 3


Determine o domínio de cada função:

a) $\displaystyle f(x)=\frac{3x+1}{x+4}$

b) $\displaystyle g(x)=\sqrt{5-x}$

c) $\displaystyle h(x)=\frac{\sqrt{x-2}}{x-5}$

d) $\displaystyle p(x)=\frac{\sqrt[3]{x-2}}{x^{2}+1}$

##### Solução do item a:


O denominador exige

$$
x+4\neq0
\quad\Longrightarrow\quad
x\neq-4.
$$

Assim,

$$
D(f)=\mathbb{R}\setminus\{-4\}.
$$

##### Solução do item b:

A raiz quadrada exige

$$
5-x\geq0
\quad\Longrightarrow\quad
x\leq5.
$$

Logo,

$$
D(g)=(-\infty,5].
$$

##### Solução do item c:

Devemos ter

$$
x-2\geq0
\quad\text{e}\quad
x-5\neq0.
$$

Portanto,

$$
D(h)=[2,5)\cup(5,+\infty).
$$

##### Solução do item d:

Considere

$$
p(x)=\frac{\sqrt[3]{x-2}}{x^2+1}.
$$

Como a raiz possui índice ímpar, o radicando pode assumir qualquer valor real. Portanto, a expressão

$$
\sqrt[3]{x-2}
$$

não impõe restrições ao domínio.

Além disso,

$$
x^2+1>0
$$

para todo $x\in\mathbb{R}$, de modo que o denominador nunca é igual a zero.

Assim, não há valores reais a serem excluídos.

Logo,

$$
D(p)=\mathbb{R}.
$$

#### Exercício 4

Determine o domínio da função

$$
f(x)=\frac{1}{x-2}+\frac{1}{x^{2}-1}.
$$

##### Solução:

Para que a função esteja definida, seus denominadores devem ser diferentes de zero.

Do primeiro denominador,

$$
x-2\neq0,
$$

obtemos

$$
x\neq2.
$$

Para o segundo denominador,

$$
x^2-1\neq0.
$$

Fatorando a diferença de dois quadrados,

$$
(x-1)(x+1)\neq0.
$$

Logo,

$$
x\neq1
\qquad\text{e}\qquad
x\neq-1.
$$

Portanto, os valores $-1$, $1$ e $2$ devem ser excluídos do domínio.

Assim,

$$
D(f)=\mathbb{R}\setminus\{-1,1,2\}.
$$

Um gráfico permite observar de forma visual como os valores de uma função se comportam.

Ao ler um gráfico, é importante distinguir:

- a coordenada horizontal, associada à entrada $x$;
- a coordenada vertical, associada à saída $y = f(x)$;
- os pontos em que $f(x)=0$;
- os trechos em que os valores da função aumentam, diminuem ou se mantêm inalterados;
- os maiores e menores valores observados;
- o domínio e a imagem representados.

Considere o gráfico da função $f$ a seguir.<br><br>

Com base no gráfico, determine:

a) $f(-4)$

b) $f(1)$

c) $f(0)$

d) o domínio da função representada;

$$
f(0)=\frac{3+0}{2}=\frac{3}{2}=1{,}5.
$$

$$
f(0)=1{,}5.
$$

d) O gráfico está definido desde $x=-4$ até $x=5$. Portanto,

$$
D(f)=[-4,6].
$$

e) O menor valor assumido pela função é $-2$ e o maior é $3$. Como o gráfico percorre todos os valores entre eles,

$$
\operatorname{Im}(f)=[-2,3].
$$

## 2.2 Zeros e Sinal de uma Função


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

<b>Zeros da função</b>

Um <b>zero da função</b> é um valor de $x$ para o qual

$$
f(x)=0.
$$

No gráfico, os zeros correspondem aos pontos em que a curva encontra o eixo horizontal.

</div>

<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

<b>Sinal da função</b>

O sinal da função pode ser identificado observando a posição do gráfico em relação ao eixo horizontal:

- se o gráfico está acima do eixo horizontal, então $f(x)>0$;
- se o gráfico está abaixo do eixo horizontal, então $f(x)<0$;
- se o gráfico está sobre o eixo horizontal, então $f(x)=0$.

</div>

#### Exemplo 2


<center> <img src="img/fig4.jpg" alt="Gráfico da função f" width="400" height="200"> </center>

Com base no gráfico, determine:

a) os zeros da função;

b) os valores de $x$ para os quais $f(x)>0$;

c) os valores de $x$ para os quais $f(x)<0$.

##### Soluções:


a) Os zeros de uma função são as abscissas dos pontos do gráfico que se encontram sobre o eixo horizontal.

$$
x=-3,\qquad x=1,\qquad x=4.
$$

Portanto, esses são os zeros da função.

b) O gráfico fica acima do eixo horizontal para

$$
x\in(-3,1)\cup(4,6].
$$

Logo,

$$
f(x)>0
\quad\text{em}\quad
(-3,1)\cup(4,6].
$$

c) O gráfico fica abaixo do eixo horizontal para

$$
x\in[-4,-3)\cup(1,4).
$$

Assim,

$$
f(x)<0
\quad\text{em}\quad
[-4,-3)\cup(1,4).
$$


## 2.3 Crescimento, Decrescimento, Máximos e Mínimos


<div style="background-color:#eef6ff; color:#FF0000; border-left:5px solid #f00b4b; padding:10px; border-radius:6px;">

<b>Crescimento, decrescimento e constância</b><br>

A leitura de crescimento, decrescimento ou constância é feita <b>da esquerda para a direita</b>.

- A função é <b>crescente</b> em um intervalo quando os valores de $f(x)$ aumentam à medida que $x$ aumenta. Em símbolos,
    
$$
x_1<x_2 \Rightarrow f(x_1)<f(x_2).
$$

- A função é <b>decrescente</b> em um intervalo quando os valores de $f(x)$ diminuem à medida que $x$ aumenta. Em símbolos,

$$
x_1<x_2 \Rightarrow f(x_1)>f(x_2).
$$

- A função é <b>constante</b> em um intervalo quando os valores de $f(x)$ permanecem iguais à medida que $x$ varia. No gráfico, isso corresponde a um trecho horizontal. Em símbolos, para $x_1$ e $x_2$ pertencentes ao intervalo,

$$
x_1<x_2 \Rightarrow f(x_1)=f(x_2).
$$

</div>

<b>Valores máximo e mínimo</b>

O gráfico também permite identificar os maiores e os menores valores assumidos pela função no trecho representado.

- O <b>valor máximo</b> é o maior valor de $f(x)$ observado no gráfico;
- o <b>valor mínimo</b> é o menor valor de $f(x)$ observado no gráfico.

Graficamente, esses valores são identificados pela coordenada vertical $y=f(x)$ dos pontos mais alto e mais baixo do gráfico, respectivamente.

## 2.4 Pontos Abertos e Fechados


Em alguns gráficos, um ponto pode aparecer **fechado (preenchido)** ou **aberto (vazio)**.

- um ponto fechado indica que o ponto pertence ao gráfico;
- um ponto aberto indica que o ponto não pertence ao gráfico.

Essa distinção é importante na leitura dos valores da função, do domínio e da imagem.

#### Exemplo 4


Considere o gráfico da função $f$ a seguir.<br><br>

<center> <img src="img/fig5.jpg" alt="Gráfico da função f" width="400" height="200"> </center>

Com base no gráfico, determine:

a) $f(1)$;

b) o domínio da função;

c) a imagem da função.

##### Solução do item a:

a) Em $x=1$, observamos um ponto fechado em $(1,2)$ e um ponto aberto em $(1,3)$.

O ponto fechado indica o valor que pertence ao gráfico para $x=1$. 

Portanto, $f(1)=2$.

##### Solução do item b:

Para determinar o domínio, observamos os valores de $x$ correspondentes aos pontos do gráfico, isto é, sua extensão ao longo do eixo horizontal.

No primeiro trecho, os valores de $x$ variam de $-2$ até $1$, incluindo os dois extremos:

$$
-2\leq x\leq1.
$$

No segundo trecho, os valores de $x$ são maiores que $1$ e vão até $3$:

$$
1<x\leq3.
$$

Embora o segundo trecho não inclua $x=1$, esse valor já pertence ao primeiro trecho. Portanto,

$$
D(f)=[-2,3].
$$

##### Solução do item c:

Para determinar a imagem, observamos os valores de $y=f(x)$ atingidos pelo gráfico, isto é, sua extensão ao longo do eixo vertical.

No primeiro trecho, os valores de $y$ variam de $-1$ até $2$, incluindo os dois extremos:

$$
-1\leq y\leq2.
$$

No segundo trecho, os valores de $y$ são maiores que $3$ e vão até $7$. O valor $y=3$ não é atingido, pois o ponto $(1,3)$ é aberto, enquanto $y=7$ é atingido. Assim,

$$
3<y\leq7.
$$

Logo,

$$
\operatorname{Im}(f)=[-1,2]\cup(3,7].
$$

Considere o gráfico da função $f$ a seguir.<br><br>

#### Exercício 6

Considere o gráfico da função $f$ a seguir.

Com base no gráfico, determine:

a) $f(-1)$;

b) $f(1)$;

c) os zeros da função;

d) o domínio da função;

e) a imagem da função.

##### Soluções:

a) Em $x=-1$, o único ponto indicado é o ponto aberto $(-1,2)$. Como esse ponto não pertence ao gráfico, $f(-1)$ não está definido.

b) Em $x=1$, também encontramos apenas um ponto aberto, agora em $(1,0)$. Portanto, $f(1)$ não está definido.

c) Os zeros da função são os valores de $x$ para os quais o gráfico possui um ponto com coordenada vertical igual a zero.

No primeiro trecho, o gráfico encontra o eixo horizontal em $x=-3$.

Além disso, existe o ponto fechado $(-2,0)$.

Portanto, os zeros da função são $x=-3$ e $x=-2$.

Observe que o ponto $(1,0)$ não fornece um zero da função, pois é aberto e não pertence ao gráfico.

d) No primeiro trecho, os valores de $x$ variam de $-4$ até $-1$, sem incluir $-1$.

No segundo trecho, os valores de $x$ são maiores que $1$ e vão até $4$.

Assim, $D(f)=[-4,-1)\cup(1,4]$.

O ponto fechado em $(-2,0)$ garante que $x=-2$ continua pertencendo ao domínio, embora o ponto $(-2,1)$ tenha sido retirado do gráfico.

e) Para determinar a imagem, observamos os valores de $y$ atingidos.

O segundo trecho assume todos os valores de $-3\leq y<0$.

O primeiro trecho fornece os valores de $y$ entre $-1$ e $2$, sem atingir $y=1$, pois o ponto $(-2,1)$ foi retirado, nem $y=2$, pois o ponto $(-1,2)$ é aberto.

Combinando os dois trechos, $\operatorname{Im}(f)=[-3,1)\cup(1,2)$.
