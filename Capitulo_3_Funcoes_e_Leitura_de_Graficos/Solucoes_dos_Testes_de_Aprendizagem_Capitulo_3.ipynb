# Teste de Aprendizagem da Aula 1

#### Questão 1

Considere a função

$$
f(x)=x^2-2x-3.
$$

Sabe-se que

$$
f(-2)+f(0)+f(a)=2.
$$

É correto afirmar que os valores de $a$ são:

a) $-3$ e $1$

b) $-1$ e $3$

c) $1$ e $3$

d) $-3$ e $-1$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Inicialmente,

$$
f(-2)=(-2)^2-2(-2)-3=5
$$

e

$$
f(0)=0^2-2\cdot0-3=-3.
$$

Assim,

$$
5-3+f(a)=2.
$$

Logo,

$$
f(a)=0.
$$

Como

$$
f(a)=a^2-2a-3,
$$

temos

$$
a^2-2a-3=0.
$$

Fatorando,

$$
(a-3)(a+1)=0.
$$

Portanto,

$$
a=-1
\qquad\text{ou}\qquad
a=3.
$$

A alternativa correta é b).

#### Questão 2

O domínio da função

$$
f(x)=\frac{\sqrt{x+1}}{x^2-2x-3}
$$

é:

a) $[-1,+\infty)$

b) $[-1,3)\cup(3,+\infty)$

c) $(-1,3)\cup(3,+\infty)$

d) $(-1,+\infty)$

e) $\mathbb{R}\setminus\{-1,3\}$

##### Solução:

Para que a função esteja definida, devemos considerar simultaneamente as restrições impostas pela raiz quadrada e pelo denominador.

Como temos

$$
\sqrt{x+1},
$$

é necessário que

$$
x+1\geq0,
$$

ou seja,

$$
x\geq-1.
$$

Além disso, o denominador deve ser diferente de zero:

$$
x^2-2x-3\neq0.
$$

Fatorando,

$$
x^2-2x-3=(x-3)(x+1).
$$

Portanto,

$$
(x-3)(x+1)\neq0,
$$

o que implica

$$
x\neq3
\qquad\text{e}\qquad
x\neq-1.
$$

Combinando as condições $x\geq-1$, $x\neq-1$ e $x\neq3$, obtemos

$$
D(f)=(-1,3)\cup(3,+\infty).
$$

Logo, a alternativa correta é c).

#### Questão 3

Considere $x$ e $y$ números reais. Qual das relações abaixo define $y$ como função de $x$ no domínio indicado?

a) $x^2+y^2=9, \ \ -3\leq x\leq3$

b) $y^2=x^2+1, \ \ x\in\mathbb{R}$

c) $x=2, \ \ y\in\mathbb{R}$

d) $y=\sqrt{9-x^2}, \ \ -3\leq x\leq3$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Para que uma relação defina $y$ como função de $x$, cada valor de $x$ do domínio deve estar associado a um único valor de $y$.

No item a),

$$
x^2+y^2=9
$$

implica

$$
y=\pm\sqrt{9-x^2}.
$$

Assim, para valores de $x$ no interior do intervalo $[-3,3]$, existem dois valores possíveis de $y$. Portanto, essa relação não define $y$ como função de $x$.

No item b),

$$
y^2=x^2+1
$$

implica

$$
y=\pm\sqrt{x^2+1}.
$$

Como $x^2+1>0$ para todo $x\in\mathbb{R}$, cada valor de $x$ está associado a dois valores distintos de $y$. Portanto, essa relação não define $y$ como função de $x$.

No item c),

$$
x=2
$$

com $y\in\mathbb{R}$ associa o mesmo valor $x=2$ a infinitos valores de $y$. Portanto, essa relação não define $y$ como função de $x$.

No item d),

$$
y=\sqrt{9-x^2},
\qquad -3\leq x\leq3,
$$

a raiz quadrada representa o valor não negativo da raiz. Assim, para cada valor de $x$ do domínio existe um único valor real de $y$.

Portanto, essa relação define $y$ como função de $x$.

Logo, a alternativa correta é d).

#### Questão 4

Considere a função

$$
f(x)=4x-3.
$$

Sabe-se que os pontos

$$
A=(2,y)
\qquad\text{e}\qquad
B=(x,13)
$$

pertencem ao gráfico de $f$.

Os valores de $y$ e $x$, respectivamente, são:

a) $5$ e $3$

b) $5$ e $5$

c) $7$ e $3$

d) $7$ e $4$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Como o ponto $A=(2,y)$ pertence ao gráfico de $f$, temos

$$
y=f(2).
$$

Assim,

$$
y=4\cdot2-3=5.
$$

Portanto,

$$
y=5.
$$

Para o ponto $B=(x,13)$, sabemos que

$$
f(x)=13.
$$

Logo,

$$
4x-3=13.
$$

Assim,

$$
4x=16
\quad\Longrightarrow\quad
x=4.
$$

Portanto, os valores de $y$ e $x$, respectivamente, são

$$
5\qquad\text{e}\qquad4.
$$

A alternativa correta é e).

#### Questão 5

Considere a função

$$
f(x)=\frac{\sqrt{ax+b}}{ax-1},
$$

em que $a$ e $b$ são números reais e $a\neq0$.

Sabendo que o domínio de $f$ é

$$
D(f)=[2,+\infty),
$$

é correto afirmar que:

a) $b=-2a$ e $a>\dfrac{1}{2}$

b) $b=2a$ e $a>\dfrac{1}{2}$

c) $b=-2a$ e $0<a<\dfrac{1}{2}$

d) $b=2a$ e $a<0$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Para que a raiz quadrada esteja definida, devemos ter

$$
ax+b\geq0.
$$

Como o domínio desejado é

$$
[2,+\infty),
$$

a condição imposta pela raiz deve ser satisfeita exatamente para $x\geq2$. Portanto, necessariamente,

$$
a>0
$$

e o valor em que o radicando se anula deve ser $x=2$.

Assim,

$$
2a+b=0,
$$

de onde

$$
b=-2a.
$$

Nesse caso,

$$
ax+b=a(x-2),
$$

e, como $a>0$,

$$
a(x-2)\geq0
\quad\Longleftrightarrow\quad
x\geq2.
$$

Entretanto, também precisamos analisar o denominador:

$$
ax-1\neq0.
$$

Portanto,

$$
x\neq\frac{1}{a}.
$$

Como queremos que **todos** os valores $x\geq2$ pertençam ao domínio, o valor

$$
\frac{1}{a}
$$

não pode pertencer ao intervalo $[2,+\infty)$.

Como $a>0$, devemos ter

$$
\frac{1}{a}<2.
$$

Multiplicando por $a>0$,

$$
1<2a,
$$

isto é,

$$
a>\frac{1}{2}.
$$

Portanto,

$$
b=-2a\qquad\text{e}\qquad a>\frac12.
$$

Logo, a alternativa correta é a).

#### Questão 2

Considere o gráfico da função $f$ a seguir.<br><br>

<center> <img src="img1/fig2.jpg" alt="Gráfico da função f" width="500"> </center>

O domínio e a imagem da função são, respectivamente:

a) $[-4,0)\cup(0,3]$ e $[-1,3)$

b) $[-4,3]$ e $[-1,3)$

c) $(-4,3)$ e $(-1,3)$

d) $[-4,0)\cup(0,3]$ e $[-1,3]$

e) Nenhuma das alternativas anteriores está correta.

##### Solução:

Para determinar o domínio, observamos os valores de $x$ para os quais existe algum ponto no gráfico.

O primeiro trecho vai de $x=-4$ até $x=0$, e o segundo vai de $x=0$ até $x=3$. Embora o ponto $(0,3)$ seja aberto, existe o ponto fechado $(0,0)$. Portanto, $x=0$ pertence ao domínio.

Assim,

$$
D(f)=[-4,3].
$$

Para determinar a imagem, observamos os valores de $y$ atingidos pelo gráfico.

O primeiro trecho fornece os valores de $-1$ até $3$, sem atingir $3$ nesse trecho. Entretanto, o segundo trecho atinge $y=3$ no ponto fechado $(3,3)$.

Logo,

$$
\operatorname{Im}(f)=[-1,3].
$$

Nenhuma das alternativas de a) a d) apresenta simultaneamente esses dois conjuntos.

Portanto, a alternativa correta é e).
